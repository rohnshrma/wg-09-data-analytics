# Collections :
used to store multiple values of similar or different data types

- list - Ordered, Indexed, changable (mutable) and allows duplicates
- tuple - Ordered, Indexed, unchangable (immutable) and allows duplicates
- dictionary - Ordered, Unindexed, changable (mutable) and no duplicates allowed
- set - unOrdered, Unindexed, unchangable (immutable) and no duplicates allowed

# List 
[ ] are used to create a list where items are seperated using `,`

```python
# A list of numbers. Some numbers repeat (4 and 23). Lists allow this.
marks = [23,34,4,23,4,4,4,23,43]
```

```python
# type() shows it is a list.
# len() counts all items. Answer: 9 (repeated numbers count too).
print(type(marks) , len(marks))
```

```python
print(marks) # ordered and allows duplicates
# Items come out in the same order we typed them. That is "ordered".
# 4 shows up many times. That is "allows duplicates".
```

```python
# indexed
# Every item has a position number (index). We use it to get the item.
print(marks[0])     # index 0 = the FIRST item
print(marks[-1])    # -1 = the LAST item
print(marks[0:4])   # slicing: takes a piece of the list
# It starts at index 0 and stops before index 4. So 4 items.
```

```python
marks[0] = 100
print(marks) # changable
# Lists can be changed. We put 100 at index 0.
# The old value (23) is replaced.
```

# list methods
Methods are ready-made tools for lists. Write them like this: marks.something()

```python
# add a new item at the end of list

marks.append(100999)
print(marks)
# append() adds ONE item at the end.
```

```python
# add a new item at the specified index

marks.insert(4 , 1999)
print(marks)
# insert(position, value) adds the item at that position.
# The other items move one step right.
```

```python
# add new items at the end of list
marks.extend( [199, 299,399] ) # another list, set, tuple
print(marks)
# extend() adds every item from the other list, one by one. We get 3 new items.
# append([199,299,399]) would add the whole list as ONE item.
```

```python
# removes and returns the item on the specified index (by default -1)
marks.pop()
marks.pop(-4)
print(marks)
# pop() removes an item and gives it back to us.
# pop() with no number removes the LAST item.
# pop(-4) removes the item 4th from the end (of the updated list).
```

```python
# removes thefirst occurance of the item provided, raises an error if item not in list
marks.remove(4)
print(marks)
# remove() looks for the VALUE 4 and removes only the FIRST one.
# pop() uses POSITION. remove() uses VALUE.
```

```python
# removes thefirst occurance of the item provided
marks.remove(1111114)
print(marks)
# ERROR on purpose. 1111114 is not in the list.
# Python gives a ValueError.
```

```python
# return number of occurances of the item specified
print(marks.count(4))
# count() tells how many times 4 is in the list.
```

```python
# return index of the first occurances of the item specified
print(marks.index(4)) # item
# index() gives the position of the FIRST 4. Counting starts from 0.

print(marks.index(4 , 5)) # item , start position(inclusive)
# The second number means: start looking from position 5.
```

```python
marks[::-1]
# [::-1] means "go through the list backwards".
# It gives a new reversed list. The original does not change.
```

```python
# modifies the list and reverse the item order
marks.reverse()
print(marks)
# reverse() flips the original list itself. It is permanent.
# Run it again to flip it back.
```

```python
marks.sort() # a-z
print(marks)
# sort() arranges the original list: small to big (numbers) or A to Z (text).
```

```python
marks.sort(reverse=True) # z-a
print(marks)
# reverse=True sorts big to small.
```

# More List Methods
The methods above are the most used. Here are the last two.

### copying a list
`copy()`

```python
original = [1, 2, 3]

# copy() makes a new, separate list with the same items.
duplicate = original.copy()

duplicate.append(4)   # we change only the copy

print(original)   # still [1, 2, 3]. The original did not change.
print(duplicate)  # only the copy has 4
```

```python
# What if we do NOT use copy()?
original2 = [1, 2, 3]
sameList = original2   # no new list. It is the SAME list with a second name.

sameList.append(4)     # we change sameList

print(original2)  # it changed too! Both names point to one list.
print(sameList)
```

### emptying a list
`clear()`

```python
toDelete = [10, 20, 30, 40]

# clear() removes every item. The list is still there, but empty.
toDelete.clear()

print(toDelete)          # []
print(len(toDelete))     # 0
```
