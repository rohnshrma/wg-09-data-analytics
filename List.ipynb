# Collections :
used to store multiple values of similar or different data types

- list - Ordered, Indexed, changable (mutable) and allows duplicates
- tuple - Ordered, Indexed, unchangable (immutable) and allows duplicates
- dictionary - Ordered, Unindexed, changable (mutable) and no duplicates allowed
- set - unOrdered, Unindexed, unchangable (immutable) and no duplicates allowed

# List 
[ ] are used to create a list where items are seperated using `,`

```python
# a list holding a bunch of numbers, some of them repeated on purpose (like 4 and 23)
# lists are totally fine with the same value showing up more than once, that's the "allows duplicates" part
marks = [23,34,4,23,4,4,4,23,43]
```

```python
# type() confirms marks is really a list
# len() counts how many items are sitting inside it right now, all 9 numbers get counted, including the repeated ones
print(type(marks) , len(marks))
```

```python
print(marks) # ordered and allows duplicates
# printing the whole thing out, notice the numbers come out in the exact same order we typed them in, that's what "ordered" means
# and 4 shows up multiple times without python complaining about it, that's the "allows duplicates" part
```

```python
# indexed
# every item in a list has a position number (called an index), and we can grab any item just by pointing at its position
print(marks[0])     # index 0 always means the very FIRST item in the list
print(marks[-1])    # negative index counts backwards from the end, so -1 always means the LAST item
print(marks[0:4])   # this is called slicing, it grabs a chunk of the list starting at index 0 up to (but NOT including) index 4, so 4 items total
```

```python
marks[0] = 100
print(marks) # changable
# lists are mutable, meaning we are allowed to reach into a specific position and swap the value sitting there, no need to rebuild the whole list
# here we replaced whatever was at index 0 (which was 23) with 100, and the list just updates itself in place
```

# list methods
these are built-in tools that already come attached to every list, you use them by writing dot-then-method-name after the list, like marks.something()

```python
# add a new item at the end of list

marks.append(100999)
print(marks)
# append() always sticks exactly ONE new item onto the very end of the list, nothing else in the list gets touched or moved around
```

```python
# add a new item at the specified index

marks.insert(4 , 1999)
print(marks)
# insert(position, value) drops the new value in at whatever position you choose, and everything that was already there from that position onward just shifts one step to the right to make room
```

```python
# add new items at the end of list
marks.extend( [199, 299,399] ) # another list, set, tuple
print(marks)
# extend() dumps every item from the other list/set/tuple onto the end of marks, one by one, so we gain 3 separate new items here
# this is different from append(), if we had done marks.append([199,299,399]) instead, it would have shoved the whole list in as a single item instead of merging the numbers in individually
```

```python
# removes and returns the item on the specified index (by default -1)
marks.pop()
marks.pop(-4)
print(marks)
# pop() is unique because it doesn't just delete an item, it also hands that removed item back to you (we're just not capturing/using it here)
# the first pop() with no number removes the very LAST item by default
# pop(-4) removes whatever item is sitting 4 positions from the end at that moment (remember, the list already changed after the first pop, so we are counting on the updated list)
```

```python
# removes thefirst occurance of the item provided, raises an error if item not in list
marks.remove(4)
print(marks)
# remove() searches for the VALUE 4 and deletes only the FIRST one it comes across, any other 4s still hanging around in the list are left untouched
# this is different from pop(), pop() removes something based on its POSITION, remove() removes something based on its VALUE
```

```python
# removes thefirst occurance of the item provided
marks.remove(1111114)
print(marks)
# this cell is here on purpose to show what happens when the value doesn't exist in the list at all
# since 1111114 was never actually one of our marks, python has nothing to delete and throws a ValueError instead of just quietly doing nothing
# (this is the "raises an error if item not in list" warning mentioned two cells above, actually happening)
```

```python
# return number of occurances of the item specified
print(marks.count(4))
# count() just goes through the whole list and tells you how many times the exact value 4 shows up in it right now
```

```python
# return index of the first occurances of the item specified
print(marks.index(4)) # item
# index() hunts for the value 4 and gives back the position of the FIRST match it finds, counting from 0 like always

print(marks.index(4 , 5)) # item , start position(inclusive)
# adding a second number tells index() to only start looking from position 5 onwards, so it skips over any matches sitting before that point
```

```python
marks[::-1]
# this slicing trick [::-1] means "give me everything, but walk through it backwards"
# it hands back a brand new reversed list without touching or reordering the original marks list at all
```

```python
# modifies the list and reverse the item order
marks.reverse()
print(marks)
# unlike the slicing trick above, reverse() actually flips the order of items INSIDE the original list itself, permanently, there's no going back unless you reverse it again
```

```python
marks.sort() # a-z
print(marks)
# sort() rearranges the items inside the list itself from smallest to biggest (for numbers) or alphabetically (for text), and it changes the original list directly
```

```python
marks.sort(reverse=True) # z-a
print(marks)
# passing reverse=True just flips the sorting direction, so now it goes from biggest to smallest instead of the usual smallest to biggest
```

# More List Methods
the methods above cover most of what you'll use day to day, here are the last two built-in list methods that hadn't shown up yet, just so this notebook covers every single one of them

### copying a list
`copy()`

```python
original = [1, 2, 3]

# copy() hands you back a brand new, completely separate list that just happens to have the same items inside it right now
duplicate = original.copy()

duplicate.append(4)   # changing the copy...

print(original)   # ...does NOT touch the original list at all, it's still just [1, 2, 3]
print(duplicate)  # only the copy grew the new item
```

```python
# just to prove why copy() actually matters: this is what happens if you DON'T copy and instead just do "sameList = original"
original2 = [1, 2, 3]
sameList = original2   # this does not make a new list, it just gives the SAME list a second name/label to be called by

sameList.append(4)     # so changing sameList...

print(original2)  # ...actually changes original2 too, because they were always pointing at the one and only same list in memory
print(sameList)
```

### emptying a list
`clear()`

```python
toDelete = [10, 20, 30, 40]

# clear() wipes out every single item in the list, leaving it completely empty, but the list itself still exists (it's not deleted, just emptied out)
toDelete.clear()

print(toDelete)          # this now prints an empty list []
print(len(toDelete))     # and its length has dropped all the way down to 0
```
