# if 
the code inside of the if block runs only when the given condition is True


```python
if condition:
    # code to be executed if the above condition is true
```

````python
# "if" checks a condition and gives True or False. 12 > 10 means "is 12 bigger than 10?" - that is True.
if 12 > 10:
    print("Greater")       # this line is INSIDE the if block (see the indent/space before it), so it runs
    print("done running")  # same indent as the line above, so this also runs, still part of the if block
````

# else
the code inside of the else block runs when all the conditions above are False

```python
if condition:
    # code to be executed if the above condition is true
else:
    # code to be executed when the condition above is False
```

````python
# input() always gives back text (called a "string"), even if the user typed numbers.
# int() converts that text into a whole number so we can compare it with >= below.
age = int(input("Enter your age : "))

# check if age is 18 or more
if age >= 18:
    print("Eligible to vote!")      # runs only when "age >= 18" is True
else:
    print("Not Eligible to vote!")  # runs when "age >= 18" is False, meaning age is less than 18
````

# elif: fallback condition
the code inside of elif block runs when the above condition in elif is True but all the conditions above are False


```python
if condition:
    # code to be executed if the above condition is true
elif condition:
    # code to be executed if the conditions above are false and this elif's condition is True
else:
    # code to be executed when all the conditions above are False
```

````python
# ask the user to type a number for today's day and turn that text into an int
today = int(input("Enter the day today (0-6) : "))

# Python checks these conditions one by one, top to bottom, and stops at the FIRST one that is True
if today == 0:
    print("Sunday")
elif today == 1:        # this line is only checked if "today == 0" turned out False
    print("Monday")
elif today == 2:
    print("Tuesday")
elif today == 3:
    print("Wednesday")
elif today == 4:
    print("Thursday")
elif today == 5:
    print("Friday")
elif today == 6:
    print("Saturday")
else:                    # runs only when NONE of the conditions above matched (e.g. user typed 99)
    print("Invalid Input")
````

````python
word = input("Enter a word : ")

# "in" checks if our word exists anywhere inside that long piece of text (like Ctrl+F, but for code)
if word in "hello world my name is john doe":
    print("Inclusive")   # runs if the word IS found somewhere inside the sentence
else:
    print("Exclusive")   # runs if the word is NOT found anywhere in the sentence
````

````python
# Real life example: convert a total number of seconds into hours, minutes and seconds
# This cell doesn't use if/else, but it uses "//" and "%" which you will often see INSIDE if conditions

seconds = 7500  # total seconds we start with (just a fixed number here, like a stopwatch reading)

hours = seconds // 3600  # "//" is floor division - it divides and drops anything after the decimal point.
                          # 7500 seconds / 3600 (seconds in an hour) = 2.08..., "//" keeps only the whole number: 2

remaining = seconds % 3600  # "%" is the modulus operator - it gives the LEFTOVER after dividing.
                             # this tells us how many seconds are left after removing the 2 full hours

minutes = remaining // 60  # from the leftover seconds, find how many full 60-second chunks (minutes) fit inside
seconds_left = remaining % 60  # whatever seconds are left after removing those full minutes

print(hours)        # how many full hours were in 7500 seconds
print(minutes)       # how many full minutes were left after removing the hours
print(seconds_left)  # how many seconds were left after removing the hours and minutes
````

# Nested if / else: an "if" placed inside another "if"

Sometimes a decision only makes sense once an earlier decision is already true - for example, "check if they have a good salary" only makes sense AFTER you already know they are an adult. To handle this, we put one if/else block INSIDE another if/else block. This is called "nesting".

Python always checks the **outer** condition first. Only when the outer condition is True does it even look at the **inner** (nested) if/else. If the outer condition is False, the inner if/else is skipped completely - it never even runs.

```python
if outer_condition:
    if inner_condition:
        # runs only when BOTH outer_condition and inner_condition are True
    else:
        # runs when outer_condition is True but inner_condition is False
else:
    # runs when outer_condition itself is False (the inner if/else is never checked at all)
```


````python
# loan eligibility program
# a bank only lends money if the person is an adult (outer check) AND, on top of that,
# they must have either a good credit score or a good salary (inner check)

good_credit_score = False  # this person's credit score is NOT good
good_salary = True         # but this person does earn a good salary
age = 31                   # this person's age

if age >= 18:                                  # OUTER if: first check whether the person is even an adult
    if good_credit_score or good_salary:        # INNER if: only checked when "age >= 18" was True. "or" means AT LEAST ONE of the two needs to be True
        print("Eligible for loan")               # runs when age >= 18 AND (good_credit_score OR good_salary) is True
    else:
        print("Not Eligible due to Financial Reasons.")  # runs when age >= 18 is True but BOTH credit score and salary are bad
else:
    print("Not Eligible for loan due to being underage")  # runs when age >= 18 is False; the inner if/else never even runs
````
