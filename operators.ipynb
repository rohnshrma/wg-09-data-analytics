# Arithmetic Operators
Operators are small symbols like `+` and `-`. They tell Python to do work on values. Arithmetic operators are for maths, like a calculator.

```python
# A variable is a box with a name. It keeps a value.
a = 23 # box "a" keeps 23
b = 3  # box "b" keeps 3
```

```python
# print() shows the answer. Text after # is only a note. Python ignores it.
print(a + b) # addition (sum): 23 + 3 = 26
print(a - b) # subtraction (difference): 23 - 3 = 20
print(a * b) # multiplication (product): 23 x 3 = 69
print(a / b) # divide (quotient): 7.666... (always with decimals)
print(a % b) # divide (remainder): 23 / 3 = 7, and 2 is left. Answer: 2
print(a ** b) # exponential (power): 23 x 23 x 23 = 12167
print(a // b) # floor division: 7.66 becomes 7 (decimal part is cut)
```

# assignment operator
`=` does not mean "equal" like in maths. It means "put the right value in the left box". Short forms like `+=` do the maths and save the answer in the same box.

```python
a = 43 # regular assignment: put 43 in box "a"
print(a)
a += 5 # ADD AND ASSIGN: a = a + 5 -> 48
print(a)
a -= 5 # sub AND ASSIGN: a = a - 5 -> 43
print(a)
a *= 5 # multi AND ASSIGN: a = a * 5 -> 215
print(a)
a /= 5 # quotient AND ASSIGN: a = a / 5 -> 43.0
print(a)
a %= 5 # Remainder AND ASSIGN: a = a % 5 -> 3.0
print(a)
a **= 5 # exponential AND ASSIGN: a = a ** 5 -> 243.0
print(a)
a //= 5 # floor division AND ASSIGN: a = a // 5 -> 48.0
print(a)
```

```python
# comparison Operators
# They ask a question. The answer is True (yes) or False (no).
```

```python
# Check both values. a is 48.0 now. b is 3.
print(a,b)
```

```python
print(a == b) # equals to: is a same as b? (== asks a question, = puts a value)
print(a != b) # not equals to: is a different from b?
print(a > b) # greater than: is a bigger than b?
print(a >= b) # greater than or equals to: is a bigger than or same as b?
print(a < b) # lesser than: is a smaller than b?
print(a <= b) # lesser than or equals to: is a smaller than or same as b?
```

# logical operators
They join two or more questions and give one answer: True or False.

- and : expects all the conditions to be True to give a Truthy answer
- or : expects at least a single condition to be True to give a Truthy answer
- not : reverse boolean value

**Example:** "I go out if it is Sunday AND it is not raining" needs both. "I eat pizza OR burger" needs only one.

```python
# "and": both sides must be True.
print(a > b and b < a) # t and t = t (both True, so True)
print(a < b and b < a) # f and t = f (one is False, so False)
print(a < b and b > a) # f and f = f (both False, so False)
```

```python
# "or": one True side is enough.
print(a > b or b < a) # t or t = t (True)
print(a < b or b < a) # f or t = t (one is True, so True)
print(a < b or b > a) # f or f = f (none is True, so False)
```

```python
# "not" flips the answer. True becomes False.
print(not True) # False
```

```python
# 2 > 5 is False. "not" flips it to True.
print(not 2 > 5) # True
```
