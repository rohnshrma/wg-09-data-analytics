# Arithmetic Operators
Operators are small symbols (like `+` or `-`) that tell Python to do some work on values. Arithmetic operators are the ones we use for maths, just like on a calculator.

```python
# Variables are like boxes with a name where we keep a value.
# We keep the number 23 in a box named "a".
a = 23
# We keep the number 3 in a box named "b".
b = 3
```

```python
# print() shows the answer on the screen. Everything after the # sign is just a note for us, Python ignores it.
print(a + b) # addition (sum): 23 + 3 = 26
print(a - b) # subtraction (difference): 23 - 3 = 20
print(a * b) # multiplication (product): 23 x 3 = 69 (the star * means "times")
print(a / b) # divide (quotient): 23 / 3 = 7.666... (division always gives a decimal answer)
print(a % b) # divide (remainder): 23 divided by 3 is 7 and 2 is left over, so the answer is 2
print(a ** b) # exponential (power): 23 x 23 x 23 = 12167 (23 multiplied 3 times)
print(a // b) # floor division: 23 / 3 is 7.66, this cuts off the decimal part and gives only 7
```

# assignment operator
The sign `=` does NOT mean "equal" like in maths. It means "put the value on the right inside the box on the left". The short forms like `+=` do the maths and save the answer back in the same box in one step.

```python
a = 43 # regular assignment: put 43 in the box "a"
print(a)
a += 5 # ADD AND ASSIGN: same as a = a + 5, so 43 + 5 = 48
print(a)
a -= 5 # sub AND ASSIGN: same as a = a - 5, so 48 - 5 = 43
print(a)
a *= 5 # multi AND ASSIGN: same as a = a * 5, so 43 x 5 = 215
print(a)
a /= 5 # quotient AND ASSIGN: same as a = a / 5, so 215 / 5 = 43.0 (division gives a decimal, so it becomes 43.0)
print(a)
a %= 5 # Remainder AND ASSIGN: same as a = a % 5, 43.0 divided by 5 leaves 3.0
print(a)
a **= 5 # exponential AND ASSIGN: same as a = a ** 5, 3.0 multiplied 5 times = 243.0
print(a)
a //= 5 # floor division AND ASSIGN: same as a = a // 5, 243.0 / 5 is 48.6, cut the decimal part = 48.0
print(a)
```

```python
# comparison Operators
# These operators ask a question about two values. The answer is always True (yes) or False (no).
```

```python
# Show the current values of both boxes so we can compare them easily.
# "a" is 48.0 now (from the last cell) and "b" is still 3.
print(a,b)
```

```python
print(a == b) # equals to: is a the same as b? (two equal signs = asking a question, one = is putting a value)
print(a != b) # not equals to: is a different from b?
print(a > b) # greater than: is a bigger than b?
print(a >= b) # greater than or equals to: is a bigger than b OR the same as b?
print(a < b) # lesser than: is a smaller than b?
print(a <= b) # lesser than or equals to: is a smaller than b OR the same as b?
```

# logical operators
Logical operators join two or more questions (conditions) together and give one final True or False answer.

- and : expects all the conditions to be True to give a Truthy answer
- or : expects at least a single condition to be True to give a Truthy answer
- not : reverse boolean value

**Real life example:** "I will go out if it is Sunday AND it is not raining" needs both things. "I will eat pizza OR burger" is happy with any one thing.

```python
# "and" is strict. Both sides must be True, otherwise the answer is False.
print(a > b and b < a) # t and t = t  (48 > 3 is True, 3 < 48 is True, both True so True)
print(a < b and b < a) # f and t = f  (48 < 3 is False, so the whole thing is False)
print(a < b and b > a) # f and f = f  (both sides are False, so False)
```

```python
# "or" is easy going. If even one side is True, the answer is True.
print(a > b or b < a) # t or t = t  (both True, so True)
print(a < b or b < a) # f or t = t  (one side is True, so True)
print(a < b or b > a) # f or f = f  (no side is True, so False)
```

```python
# "not" flips the answer. True becomes False and False becomes True.
print(not True) # prints False
```

```python
# Python first solves 2 > 5 which is False, and then "not" flips it to True.
print(not 2 > 5) # prints True
```
