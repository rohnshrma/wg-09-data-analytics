# Variable 

its called a variable due to its name of its value being changable

### How to create a variable ?
the moment you assign a value to a word using the `=` sign it becomes a variable

```python
# This makes a variable.
# "age" is the name. 23 is the value.
# Think of a box with the label "age". The number 23 is inside.
age = 23
```

```python
# Show the value inside "age".
print(age)

# Python does 23 * 10 = 230 and shows it.
# "age" does not change. It is only a quick calculation.
print(age * 10)
```

```python
# A variable can be changed any time.
# The old value 23 is gone. Now "age" is 46.
age = 46
```

```python
# It shows 46, because we changed it above.
print(age)
```

# variable naming conventions

### Rules
- No Space
- Cannot start a variable name with a number but can contain or end with a number
- No Special symbol (start, end or between), except `_`
- DO Not use Reserved Keyword

table of reserved keywords

## Suggestions
- Start variable name with a lowercase letter
- In case of multiword variable name e.g., `helloworldmynameis`
    - `helloWorldMyNameIs` : camelCasing
    - `hello_world_my_name_is` : snake_casing 

# Data Types
- Int (integer) : number without decimal point
- Float : number with decimal point
- Str (string) : sequence of character(`Alphabets, Numbers, Special symbols or Whitespace`) enclosed in single `' '` or double `" "` quotes
- Bool / Boolean : True | False
- None : Intentional Absence of value

### Collection
- List
- Tuple
- Set
- Dictionary

```python
# Text must be inside quotes.
myname = "rohan"
```

```python
# Show the value inside "myname".
print(myname)
```

```python
# None means "nothing" on purpose.
# It is not 0. It is not an empty string "". It means no value yet.
nextPm = None
```

```python
# Values with commas and no brackets make a tuple.
# A tuple keeps many values together. You cannot change it later.
x = 23,23,45
```

```python
# type() tells the kind of data.
# x is a tuple, so it shows <class 'tuple'>.
print(type(x))
```
