# Introduction
- Output : `print()` function is used to display info on screen
- Input : `input()` function is used to collect info from the user
- Variables : empty containers used to store data having a label on them known as `variable name` or `identifier`

# output

```python
# print() is basically how python "talks" to us or shows stuff on screen
# here python first solves the math (2 + 2 = 4) and then prints the answer
print(2 + 2)
```

```python
# this time we are not doing any math, we are just printing plain text
# whatever is written inside the quotes " " is treated as text (a string), so it prints exactly as it is
print("Hello sanjay")
```

# input

```python
# input() pauses the program and waits for the user to type something and hit enter
# the text inside the brackets is just the message shown to the user, it is not stored anywhere
# whatever the user types is what input() gives back, since we are not saving it in a variable here it just shows up as output
input("Enter your name : ")
```

# Variable 

its called a variable due to its name of its value being changable

### How to create a variable ?
the moment you assign a value to a word using the `=` sign it becomes a variable

```python
# this is how we create a variable
# "age" is the name/label we are giving, and 23 is the value we are storing inside it
# think of it like a labelled box where we are keeping the number 23
age = 23
```

```python
# printing the value that is currently stored inside the variable "age"
print(age)

# python first works out age * 10 (23 * 10 = 230) and then prints the result
# note, this does not actually change the value stored in "age", it is just a calculation done on the spot
print(age * 10)
```

```python
# variables can be changed / overwritten any time we want
# here we are simply giving "age" a brand new value, the old value (23) is gone now, replaced by 46
age = 46
```

```python
# printing age again, this will now show 46 because we just updated it in the step above
print(age)
```

# variable naming conventions

### Rules
- No Space
- Cannot start a variable name with a number but can contain or end with a number
- No Special symbol (start, end or between), except `_`
- DO Not use Reserved Keyword

table of reserved keywords

## Suggestions
- Start variable name with a lowercase letter
- In case of multiword variable name e.g., `helloworldmynameis`
    - `helloWorldMyNameIs` : camelCasing
    - `hello_world_my_name_is` : snake_casing 

# Data Types
- Int (integer) : number without decimal point
- Float : number with decimal point
- Str (string) : sequence of character(`Alphabets, Numbers, Special symbols or Whitespace`) enclosed in single `' '` or double `" "` quotes
- Bool / Boolean : True | False
- None : Intentional Absence of value

### Collection
- List
- Tuple
- Set
- Dictionary

```python
# creating a variable called "myname" and storing a piece of text inside it
# since it is text, it has to be wrapped in quotes
myname = "rohan"
```

```python
# printing whatever value is currently stored inside "myname"
print(myname)
```

```python
# None is a special value in python that basically means "nothing" on purpose
# it is different from 0 or an empty string "", it literally means there is no value here yet
nextPm = None
```

```python
# when you write values separated by commas without any brackets around them, python automatically groups them into a tuple
# a tuple is just a bunch of values kept together as one unit, and once it is made you cannot change it (it is immutable)
x = 23,23,45
```

```python
# type() tells us what kind of data type a variable is holding
# since x was made using comma separated values, python stored it as a tuple, so this prints <class 'tuple'>
print(type(x))
```
