# Tuple

- `( )` are used to represent tuple
- ordered, indexed, unchangable (immutable), allows duplicates
- `( )` can be used to create a tuple
- in order to create a tuple put mulitple values divided by `,` in a single variable (with or without ` ( ) `)
- in order to create a tuple with a single item put `,` after the item (with or without ` ( ) `)

```python
# just wrapping a single string in ( ) does NOT make it a tuple, python just treats the extra brackets as nothing special
names = ("hello")
print(names , type(names))   # notice the type is still str, not tuple, because there's no comma inside

# you don't even need the round brackets ( ) to make a tuple, just separating values with commas is enough
marks = 12,34,45,44
print(marks , type(marks))   # python automatically packs these comma separated values into a tuple behind the scenes

# to make a tuple with only ONE item inside, a trailing comma after the item is a MUST
genders = "male",
print(genders, type(genders))   # without that comma at the end, python would've just seen this as a plain string, not a tuple
```

```python
# this time we're using round brackets ( ) around the comma separated values
# this is the more common and more readable way to write a tuple
marks = (12,34,24,34,4,32,2,423)
```

```python
print(type(marks) , marks) # ordered and allows duplicates
# type() confirms marks really is a tuple
# "ordered" means the items always come out in the exact same sequence we typed them in
# "allows duplicates" means numbers like 34 and 4 can each show up more than once, and python is totally fine with that
```

```python
print(marks[1])  #indexed
# every item in a tuple has a position number (an index), starting from 0
# marks[1] means "give me the item sitting at position 1", which is the SECOND item in the tuple (34)

print(marks[ 0 : 3])  #indexed
# this is called slicing, it grabs a chunk of the tuple
# starting at index 0 up to (but NOT including) index 3, so we get 3 items back
# the result of slicing a tuple is itself a tuple
```

```python
marks[1] = 340 # tuples are unchangable (immutable), we are NOT allowed to swap out an item once the tuple is made
# this line is here on purpose to trigger the error below and prove that point
# unlike a list, there is no way to reach into a tuple and replace one of its values, python simply refuses and throws an error instead
```

# tuple methods
since tuples can't be changed once made, they only come with a couple of built-in tools (nowhere near as many as lists have), here are the two most commonly used ones

```python
# index() hunts down the item you give it and tells you the position of the FIRST time it shows up in the tuple
print(marks.index(34)) # item

# giving a second number tells python to only start searching from that position onwards, skipping anything before it
print(marks.index(34 , 2)) # item ,  start position(inclusive)
```

```python
# count() goes through the whole tuple and tells you how many times the exact value shows up in it
print(marks.count(34)) # item

# if the value isn't in the tuple at all, count() doesn't throw an error, it just calmly returns 0
print(marks.count(340000)) # item
```

### editing a tuple anyway (the workaround)
since a tuple itself can't be edited directly, the usual trick is: turn it into a list first, make all the changes you need using normal list methods, then turn it back into a tuple to "lock" it again

```python
print(marks , type(marks))   # this is still our original, unchangeable tuple

marksList = list(marks)   # list() converts the tuple into a brand new list, and now editing is allowed
print(marksList , type(marksList) , type(marks))   # same values inside, but marksList is now a list while marks is still a tuple

marksList.append(10000)   # lists allow this, so we can add a new item at the end
marksList.pop(1)          # and remove the item sitting at position 1
marksList[0] = 100        # and directly overwrite the item at position 0
print(marksList)          # here's the list after all three of those changes

marks = tuple(marksList)   # tuple() converts it back into a tuple, "locking" it again so it can't be edited anymore
print(marks , type(marks))   # marks is a tuple again, but now holding the updated values that came from marksList
```

```python
# a fresh, simple tuple with exactly 3 items, just to make unpacking easy to follow
marks = (12,23,45)
```

# tuple unpacking
unpacking means pulling every item out of a tuple straight into its own separate variable, all in a single line, instead of grabbing items one at a time with indexing

```python
# the usual way, grabbing one item at a time using its index position
print(marks[0])
print(marks[1])
print(marks[2])
```

```python
# unpacking lets us skip the indexing entirely and grab ALL items in a single line
# python matches each variable on the left to the item sitting in the same position on the right, in order
rohan, sanjay, vinay = marks
print(rohan)     # gets marks[0] -> 12
print(sanjay)    # gets marks[1] -> 23
print(vinay)     # gets marks[2] -> 45
```

```python
# unpacking only works when the number of variables on the left EXACTLY matches the number of items in the tuple
# here we only gave 2 variables for a tuple that has 3 items, so python has no idea where the 3rd value should go
rohan, vinay = marks
print(rohan)
print(sanjay)
print(vinay)
# this line is here on purpose to show the error below
# python stops right at the unpacking line itself, so none of these print lines ever actually run
```

```python
# putting a * before a variable name tells python "dump whatever items are left over into this one, as a list"
# this is handy when you don't know (or don't care about) the exact number of items ahead of time
*rohan, vinay = marks
print(rohan)   # collects everything except the last item -> [12, 23]
print(vinay)   # gets just the very last item -> 45
```
