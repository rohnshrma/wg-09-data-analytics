# Tuple

- `( )` are used to represent tuple
- ordered, indexed, unchangable (immutable), allows duplicates
- `( )` can be used to create a tuple
- in order to create a tuple put mulitple values divided by `,` in a single variable (with or without ` ( ) `)
- in order to create a tuple with a single item put `,` after the item (with or without ` ( ) `)

```python
# Only ( ) around one text does NOT make a tuple.
names = ("hello")
print(names , type(names))   # type is str, because there is no comma

# You do not need ( ). Commas are enough.
marks = 12,34,45,44
print(marks , type(marks))   # Python makes a tuple

# For a tuple with ONE item, you MUST put a comma after it.
genders = "male",
print(genders, type(genders))   # without the comma it would be a string
```

```python
# This time we use ( ) around the values.
# This is the most common way to write a tuple.
marks = (12,34,24,34,4,32,2,423)
```

```python
print(type(marks) , marks) # ordered and allows duplicates
# type() shows it is a tuple.
# "ordered": items stay in the same order.
# "allows duplicates": 34 and 4 can repeat.
```

```python
print(marks[1])  #indexed
# Every item has a position number (index). It starts from 0.
# marks[1] is the SECOND item (34).

print(marks[ 0 : 3])  #indexed
# Slicing takes a piece. It starts at 0 and stops before 3. So 3 items.
# The result is also a tuple.
```

```python
marks[1] = 340 # tuples are unchangable (immutable)
# ERROR on purpose.
# A tuple can not be changed after it is made. Python refuses.
```

# tuple methods
A tuple can not be changed, so it has only two methods. Here they are.

```python
# index() gives the position of the FIRST match.
print(marks.index(34)) # item

# The second number means: start looking from that position.
print(marks.index(34 , 2)) # item ,  start position(inclusive)
```

```python
# count() tells how many times a value is in the tuple.
print(marks.count(34)) # item

# If the value is not there, it gives 0. No error.
print(marks.count(340000)) # item
```

### editing a tuple anyway (the workaround)
A tuple can not be edited. So we do this: change it to a list, edit the list, then change it back to a tuple.

```python
print(marks , type(marks))   # still the old tuple

marksList = list(marks)   # list() makes a list from the tuple. Now we can edit.
print(marksList , type(marksList) , type(marks))   # marksList is a list. marks is a tuple.

marksList.append(10000)   # add an item at the end
marksList.pop(1)          # remove the item at position 1
marksList[0] = 100        # change the item at position 0
print(marksList)          # the list after 3 changes

marks = tuple(marksList)   # tuple() makes it a tuple again (locked)
print(marks , type(marks))   # a tuple again, with the new values
```

```python
# A simple tuple with 3 items. Easy for unpacking.
marks = (12,23,45)
```

# tuple unpacking
Unpacking means taking every item out of a tuple into its own variable, in one line.

```python
# The normal way: one item at a time with the index.
print(marks[0])
print(marks[1])
print(marks[2])
```

```python
# Unpacking: get ALL items in one line.
# Each variable on the left gets the item in the same position on the right.
rohan, sanjay, vinay = marks
print(rohan)     # marks[0] -> 12
print(sanjay)    # marks[1] -> 23
print(vinay)     # marks[2] -> 45
```

```python
# The number of variables must MATCH the number of items.
# Here we have 2 variables for 3 items. Python does not know what to do.
rohan, vinay = marks
print(rohan)
print(sanjay)
print(vinay)
# ERROR on purpose.
# Python stops at the unpacking line. The prints never run.
```

```python
# A * before a name means: put all the leftover items here, as a list.
# It helps when you do not know how many items there are.
*rohan, vinay = marks
print(rohan)   # everything except the last item -> [12, 23]
print(vinay)   # the last item -> 45
```
