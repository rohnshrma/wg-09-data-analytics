# Sets
{ } or set() are used to create a new set 

- unordered, unindexed, unchangable (immutable), no duplicates are allowed

```python
# { } make a set. [ ] make a list. ( ) make a tuple.
# We typed 6 five times on purpose.
marks = {1,2,3,4,5,6,6,6,6,6}
```

```python
print(marks , type(marks) , len(marks)) # unordered and no duplicates allowed
# len() shows only 6 items. A set keeps one copy of each value.
# "unordered": items may print in a different order than we typed.
```

```python
# A set of names. Each name is only once.
names = {"sanjay","rohan","vinay"}
```

```python
print(names , marks)
# The print order is not the order we typed.
```

```python
print(names[0]) # unindexed
# ERROR on purpose.
# A set has no position numbers. There is no index 0.
```

```python
names[0] = "digvijay" # immutable (unchangable)
# ERROR on purpose.
# A set has no positions, so we can not change an item by position.
```

# Set Methods
Methods are ready-made tools. Write them like this: `marks.something()`

First we add and remove one item at a time. Later we compare TWO sets (union, intersection, difference).

```python
# add a new item in the set at a random position
# add() puts ONE new item in the set.
# The item can show up anywhere. It is not always at the end.
marks.add(100)
print(marks)
names.add("Chaman")
print(names)
```

```python
# removes a random item from the set
# pop() removes ONE item. We can not choose which one.
# A set has no order, so Python picks any item.
marks.pop()
print(marks)
```

```python
# removes  item from the set, raises an error if item not found
# remove() deletes one item by its VALUE. (A set has no positions.)
marks.remove(100)
print(marks)
```

```python
# ERROR on purpose. remove() on an item that is not in the set.
marks.remove(100)  # 100 was already removed above
print(marks)
# Python gives a KeyError.
```

```python
# removes item from the set, does nothing if item not found
# discard() also deletes an item by value.
# Difference: if the item is not there, discard() gives NO error.
marks.discard(100)
print(marks)
```

### clear() - make a set empty

```python
numbers = {1, 2, 3, 4, 5}

numbers.clear()   # clear() removes every item
print(numbers)         # set() means an empty set
# Note: {} is an empty DICTIONARY, not a set. That is why an empty set shows as set().
print(type(numbers))   # still a set, only empty
```

### copy() - make a separate copy

```python
original = {1, 2, 3}

# copy() makes a new, separate set with the same items.
duplicate = original.copy()

duplicate.add(4)   # we change only the copy

print(original)    # still {1, 2, 3}
print(duplicate)    # only the copy has 4

# Without .copy(), "sameSet = original" gives the SAME set a second name.
# Changing one would change both.
```

# Comparing two sets
Example: two friends and their hobbies are two sets. These methods find what is common, what is different, or join everything. It is like the sets in maths class.

```python
# Two hobby sets. We use them in all the examples below.
# "cricket" and "reading" are in both.
rohanHobbies = {"gaming", "coding", "cricket", "reading"}
sanjayHobbies = {"cricket", "reading", "painting", "singing"}
print(rohanHobbies)
print(sanjayHobbies)
```

```python
# union() joins BOTH sets into a new set.
# A shared item shows only ONCE.
allHobbies = rohanHobbies.union(sanjayHobbies)
print(allHobbies)

# The | symbol does the same thing. It is shorter.
print(rohanHobbies | sanjayHobbies)
```

```python
# update() is like union(), but it changes the original set. It does not make a new one.
rohanHobbiesCopy = rohanHobbies.copy()   # we use a copy to keep the original safe
rohanHobbiesCopy.update(sanjayHobbies)
print(rohanHobbiesCopy)   # now it also has sanjay's hobbies
```

```python
# intersection() finds items that are in BOTH sets. It means "what is common".
commonHobbies = rohanHobbies.intersection(sanjayHobbies)
print(commonHobbies)   # cricket and reading

# The & symbol does the same thing.
print(rohanHobbies & sanjayHobbies)
```

```python
# intersection_update() finds common items too, but changes the original set. It keeps ONLY the common items.
rohanHobbiesCopy2 = rohanHobbies.copy()
rohanHobbiesCopy2.intersection_update(sanjayHobbies)
print(rohanHobbiesCopy2)   # only the common items are left
```

```python
# difference() finds items in the FIRST set but NOT in the second. It means "only mine".
onlyRohan = rohanHobbies.difference(sanjayHobbies)
print(onlyRohan)   # rohan's hobbies that sanjay does NOT have

onlySanjay = sanjayHobbies.difference(rohanHobbies)
print(onlySanjay)  # sanjay's hobbies that rohan does NOT have
# The results are different. ORDER matters. Swap the sets and the answer swaps.

# The - symbol does the same thing.
print(rohanHobbies - sanjayHobbies)
print(sanjayHobbies - rohanHobbies)
```

```python
# difference_update() is like difference(), but it changes the original set.
rohanHobbiesCopy3 = rohanHobbies.copy()
rohanHobbiesCopy3.difference_update(sanjayHobbies)
print(rohanHobbiesCopy3)   # only hobbies that sanjay does not have
```

```python
# symmetric_difference() finds items that are in only ONE set (not in both).
uniqueToEachOther = rohanHobbies.symmetric_difference(sanjayHobbies)
print(uniqueToEachOther)   # "onlyRohan" and "onlySanjay" together

# The ^ symbol does the same thing.
print(rohanHobbies ^ sanjayHobbies)
```

```python
# symmetric_difference_update() does the same, but changes the original set.
rohanHobbiesCopy4 = rohanHobbies.copy()
rohanHobbiesCopy4.symmetric_difference_update(sanjayHobbies)
print(rohanHobbiesCopy4)
```

```python
weekendHobbies = {"cricket", "reading"}

# issubset() checks: is EVERY item of weekendHobbies inside rohanHobbies? Gives True or False.
print(weekendHobbies.issubset(rohanHobbies))   # True
```

```python
# issuperset() is the opposite of issubset().
# It checks: does this set have every item of the other, smaller set?
print(rohanHobbies.issuperset(weekendHobbies))   # True
```

```python
travelHobbies = {"trekking", "backpacking"}

# isdisjoint() checks: do the two sets have NOTHING in common?
# True = nothing common. False = at least one common item.
print(rohanHobbies.isdisjoint(travelHobbies))   # True, nothing common
print(rohanHobbies.isdisjoint(sanjayHobbies))    # False, they share cricket and reading
```
