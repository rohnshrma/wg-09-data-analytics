# Sets
{ } or set() are used to create a new set 

- unordered, unindexed, unchangable (immutable), no duplicates are allowed

```python
# { } curly braces create a set, just like [ ] makes a list and ( ) makes a tuple
# notice we typed the number 6 five times in a row on purpose
marks = {1,2,3,4,5,6,6,6,6,6}
```

```python
print(marks , type(marks) , len(marks)) # unordered and no duplicates allowed
# even though we typed 6 five times above, len() shows only 6 items total
# a set automatically throws away every extra duplicate the moment it's created, keeping just one copy of each value
# "unordered" means don't expect the items to print back out in the same sequence you typed them in, python is free to arrange them however it wants internally
```

```python
# a set of names, each name only shows up once here because that's just how we typed it (no duplicates typed this time)
names = {"sanjay","rohan","vinay"}
```

```python
print(names , marks)
# printing both sets, notice the ORDER items come out in has nothing to do with the order we originally typed them in
```

```python
print(names[0]) # unindexed
# sets don't keep track of "positions" at all, there's no index 0, index 1, and so on
# this line is here on purpose to trigger the error below and prove that point
```

```python
names[0] = "digvijay" # immutable (unchangable)
# since sets have no positions/indexes at all, there's no "index 0" to even point at and overwrite in the first place
# this line is here on purpose to trigger the error below and prove that point
```

# Set Methods
sets come with their own bunch of built-in tools, use them by writing dot-then-method-name after the set, like `marks.something()`

some methods add/remove one item at a time (right below), and later on there's a whole other group of methods for comparing TWO sets against each other (union, intersection, difference, etc.) the same way you'd compare two groups of things in real life

```python
# add a new item in the set at a random position
# add() drops exactly one new item into the set
# since sets are unordered, that new item can end up showing up ANYWHERE in the printed result, not necessarily at the end (unlike list's append())
marks.add(100)
print(marks)
names.add("Chaman")
print(names)
```

```python
# removes a random item from the set
# pop() removes and returns ONE item, but which item gets removed is NOT something we get to choose
# since sets have no order or index, python just picks whichever item it finds first internally, which can feel random to us
marks.pop()
print(marks)
```

```python
# removes  item from the set, raises an error if item not found
# remove() deletes ONE specific item that we point out by its VALUE (there's no position to point at here, remember, sets are unindexed)
marks.remove(100)
print(marks)
```

```python
# this line is here on purpose to show what happens when remove() is asked to delete something that isn't in the set
marks.remove(100)  # 100 was already removed in the cell above, so it no longer exists inside marks
print(marks)
# python has nothing to delete and throws a KeyError instead of quietly doing nothing
```

```python
# removes item from the set, does nothing if item not found
# discard() also deletes an item by its value, just like remove()
# but the big difference is: if the item isn't found, discard() stays quiet and does nothing instead of throwing an error
marks.discard(100)
print(marks)
```

### clear() - empty out a set completely

```python
numbers = {1, 2, 3, 4, 5}

numbers.clear()   # clear() wipes out every single item in the set, leaving it completely empty
print(numbers)         # this prints set(), an empty set
# note: printing {} would actually mean an empty DICTIONARY in python, not a set, that's why an empty set looks like set() instead
print(type(numbers))   # it's still a set, just an empty one, clear() emptied it out but did NOT delete the set itself
```

### copy() - make a separate, independent copy

```python
original = {1, 2, 3}

# copy() hands back a brand new, completely separate set that just happens to hold the same items right now
duplicate = original.copy()

duplicate.add(4)   # changing the copy...

print(original)    # ...does NOT touch the original set at all, it's still just {1, 2, 3}
print(duplicate)    # only the copy grew the new item

# this matters for the exact same reason it mattered for lists: without .copy(), doing "sameSet = original"
# would NOT create a new set, it would just give the SAME set a second name, so changing one would change both
```

# Comparing two sets
real life example: think of two friends and their hobbies as two sets, these methods let you find what they have in common, what's unique to each of them, or combine everything together, exactly like the "sets" you may have studied in a maths class (union, intersection, etc.)

```python
# two sets of hobbies we'll reuse for every example below, notice "cricket" and "reading" are common to both
rohanHobbies = {"gaming", "coding", "cricket", "reading"}
sanjayHobbies = {"cricket", "reading", "painting", "singing"}
print(rohanHobbies)
print(sanjayHobbies)
```

```python
# union() combines items from BOTH sets into one brand new set
# anything that exists in both only shows up ONCE in the result (no duplicates, remember)
allHobbies = rohanHobbies.union(sanjayHobbies)
print(allHobbies)

# the | symbol does the exact same job as union(), just a shorter way to write it
print(rohanHobbies | sanjayHobbies)
```

```python
# update() does basically the same job as union(), but instead of GIVING BACK a brand new set, it directly changes the original set itself
rohanHobbiesCopy = rohanHobbies.copy()   # working on a copy so we don't disturb rohanHobbies for the examples still coming up below
rohanHobbiesCopy.update(sanjayHobbies)
print(rohanHobbiesCopy)   # rohanHobbiesCopy has now permanently changed to include sanjay's hobbies too
```

```python
# intersection() finds only the items that show up in BOTH sets, basically "what do we have in common"
commonHobbies = rohanHobbies.intersection(sanjayHobbies)
print(commonHobbies)   # cricket and reading, the two hobbies both friends share

# the & symbol does the exact same job as intersection()
print(rohanHobbies & sanjayHobbies)
```

```python
# intersection_update() also finds the common items, but instead of returning a new set, it changes the original set to keep ONLY those common items
rohanHobbiesCopy2 = rohanHobbies.copy()
rohanHobbiesCopy2.intersection_update(sanjayHobbies)
print(rohanHobbiesCopy2)   # rohanHobbiesCopy2 now only contains whatever was common between the two original sets
```

```python
# difference() finds items that are in the FIRST set but NOT in the second one, basically "what's unique to me"
onlyRohan = rohanHobbies.difference(sanjayHobbies)
print(onlyRohan)   # rohan's hobbies that sanjay does NOT have

onlySanjay = sanjayHobbies.difference(rohanHobbies)
print(onlySanjay)  # sanjay's hobbies that rohan does NOT have
# notice this result is different from the one above, ORDER matters for difference(), swapping the sets swaps the answer

# the - symbol does the exact same job as difference()
print(rohanHobbies - sanjayHobbies)
```

```python
# difference_update() finds "what's unique to me" too, but changes the original set directly instead of returning a brand new one
rohanHobbiesCopy3 = rohanHobbies.copy()
rohanHobbiesCopy3.difference_update(sanjayHobbies)
print(rohanHobbiesCopy3)   # rohanHobbiesCopy3 now only keeps the hobbies that sanjay doesn't also have
```

```python
# symmetric_difference() finds everything that is NOT common, items that belong to only ONE of the two sets (either one, just not both)
uniqueToEachOther = rohanHobbies.symmetric_difference(sanjayHobbies)
print(uniqueToEachOther)   # basically "onlyRohan" and "onlySanjay" combined together

# the ^ symbol does the exact same job as symmetric_difference()
print(rohanHobbies ^ sanjayHobbies)
```

```python
# symmetric_difference_update() does the same "not common" comparison, but changes the original set directly instead of returning something new
rohanHobbiesCopy4 = rohanHobbies.copy()
rohanHobbiesCopy4.symmetric_difference_update(sanjayHobbies)
print(rohanHobbiesCopy4)
```

```python
weekendHobbies = {"cricket", "reading"}

# issubset() checks if EVERY item inside weekendHobbies also exists inside rohanHobbies, gives back True or False
print(weekendHobbies.issubset(rohanHobbies))   # True, both "cricket" and "reading" are already inside rohanHobbies
```

```python
# issuperset() is just the flipped version of issubset()
# it checks if THIS set contains every item belonging to the other, smaller set
print(rohanHobbies.issuperset(weekendHobbies))   # True, same relationship as above, just checked from the other direction
```

```python
travelHobbies = {"trekking", "backpacking"}

# isdisjoint() checks if two sets have absolutely NOTHING in common
# it gives True if there's zero overlap at all, and False if they share even a single item
print(rohanHobbies.isdisjoint(travelHobbies))   # True, none of rohan's hobbies match travelHobbies at all
print(rohanHobbies.isdisjoint(sanjayHobbies))    # False, they DO share some hobbies (cricket, reading)
```
