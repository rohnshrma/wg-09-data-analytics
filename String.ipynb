##### `type()` function return the data type of the variable / value specified

# Str (string) 
sequence of character(`Alphabets, Numbers, Special symbols or Whitespace`) enclosed in single `' '` or double `" "` quotes

```python
# even though 345 looks like a number, because it is wrapped in quotes " " python treats it as text (a string), not a number
x = "345"

# checking the data type of "123" directly, this shows str because of the quotes around it
print(type("123"))

# checking the data type of the variable x, this also shows str for the exact same reason
print(type(x))
```

```python
# just storing a name inside a variable, nothing fancy here
# python figures out on its own that this is a string because it is inside quotes
myName = "john das"
```

#### `len()` function is used to get the number of characters in a string

```python
# len() counts how many characters are in the string, spaces get counted too
# "john das" has 8 characters, including the space between john and das
print(len(myName))
```

##### Note : Value collected using the input function is of type `Str` by default, but can be converted into supported data type

using function like str, float, int , list, tuple, set, ... etc

```python
# input() always gives back text (a string) no matter what the user types, even if it looks like a number
# so we wrap it with int() here to convert that text into an actual whole number that we can do maths on
n = int(input("Enter a number : "))

# printing the value along with its data type, this should now show it as int, not str
print(n, type(n))

# since n is now a proper number, n + n does real addition (like 23 + 23 = 46)
# if we had skipped the int() conversion, this would have just stuck two pieces of text together instead of adding numbers
print(n + n)
```

# concatenation :
add two or more string together using the `+` operator

```python
# two separate variables holding two separate pieces of text
fName = "Sanjay"
lName = "Jangra"

# the + operator between strings does not add numbers, it just joins/sticks the pieces of text together, this is called concatenation
# we are manually putting spaces (" ") in between so the words do not end up jammed together with no gap
print("My name is " + fName + " " +lName)
```

```python
# STRING METHODS
# these are just built-in tools that already come attached to every string, you use them by writing dot-then-method-name after a string, like txt.something()
```

```python
# one sample sentence we will keep reusing below to test out different string methods on the same text
txt = "hello world my name is john doe"
```

```python
print(txt.index("o"))   # index() hunts for the letter "o" and tells you the position where it shows up FIRST (counting from 0, left to right)
print(txt.rindex("o"))  # rindex() does the same hunt but from the other end, so it tells you the position of the LAST "o" in the string
```

```python
# strip() chops off any blank spaces sitting at the very start and the very end of the string, but leaves spaces in the middle untouched
# note: this does not change the original string, it just gives back a brand new cleaned up copy (strings can never be changed in place in python)
"       rohan sharma     ".strip()  # remove leading and trailing
```

```python
# lstrip() only cleans up the "left" side (the beginning) of the string, whatever spaces are hanging around at the end are left alone
"       rohan sharma     ".lstrip()  # remove leading whitespace
```

```python
# rstrip() is the mirror opposite of lstrip(), it only cleans the "right" side (the end) and keeps the starting spaces as they are
"       rohan sharma     ".rstrip()  # remove trailing whitespace
```

```python
# just printing txt again here to prove none of the methods above actually touched or changed our original txt variable
# every string method we use always returns a new string, the old one stays exactly as it was
txt
```

```python
# split() chops a sentence into a list of separate words, by default it breaks the string wherever it finds whitespace (space, tab, newline)
# splits the string into a list (on whitespace by default)
print(txt.split())
```

```python
# you can also tell split() exactly which character to break on instead of the default whitespace
# splits the string into a list (on whitespace by default)
print(txt.split("a"))

# maxsplit limits how many times the split actually happens
# maxsplit=1 means "only cut once", so it stops after the very first "m" it finds and leaves the rest of the string as one single piece
print(txt.split("m" , maxsplit=1))
```

```python
# format() lets you build a sentence with placeholders {} that get filled in later with whatever values you pass in, in the same order
print("hello world my name is {} and i am an {}".format("john" , "alien"))
```

```python
# format_map() does basically the same job as format(), except instead of passing values one by one, you hand it a dictionary
# and it matches each {key} in the string to the matching key in that dictionary
info = {"name": "john", "creature": "alien"}
print("hello world my name is {name} and i am an {creature}".format_map(info))
```

```python
# splits the string into a list (on whitespace by default)
print(txt.split("m" , maxsplit=1))   # split() counts its cuts starting from the LEFT side of the string
print(txt.rsplit("m" , maxsplit=1))  # rsplit() counts its cuts starting from the RIGHT side of the string, so with maxsplit=1 it only cuts the LAST "m" it finds
```

```python
# removeprefix() deletes a piece of text ONLY if it is sitting right at the very start of the string, otherwise it leaves the string untouched
print(txt.removeprefix("hello"))  # "hello" really is at the start, so it gets removed
print(txt.removeprefix("john"))   # "john" is NOT at the start (it is in the middle somewhere), so nothing gets removed, we get the full string back unchanged

# removesuffix() is the same idea but checks the very END of the string instead of the start
print(txt.removesuffix("hello"))  # "hello" is not at the end, so nothing changes
print(txt.removesuffix("doe"))    # "doe" really is the last word, so it gets chopped off
```

```python
print(txt.split("my"))       # split() breaks the string EVERY time it sees "my" and throws away "my" itself, giving back a list of the leftover pieces
print(txt.partition("my"))   # partition() only looks for the FIRST "my" and gives back a tuple of exactly 3 things: (everything before it, "my" itself, everything after it)

# difference between split and partition
# split can chop a string into many pieces and doesn't keep the separator, partition always gives exactly 3 pieces and keeps the separator in the middle
```

```python
print(len(txt))              # just a reminder of how long our sentence actually is, so the center() call below makes more sense
print(txt.center(41))        # center() pads the string with spaces on both sides so the total length becomes 41, keeping the text in the middle
print(txt.center(41 , "-"))  # new length (more than the actual length) ,  fill char
# same idea as above, but instead of padding with plain spaces, it pads with the "-" character we gave it
```

```python
print(txt.zfill(41)) # new length (more than the actual length
# zfill() pads the string on the LEFT ONLY, and always fills the gap with zeros "0" instead of spaces
# it's mainly meant for numbers stored as text, like turning "7" into "007" so everything lines up nicely
```

```python
# join() is basically the reverse of split(), it takes a list of strings and glues them back together into one single string
# whatever string you call .join() on becomes the "glue" placed in between every item of the list
print(("-").join(['hello', 'world', 'my', 'name', 'is', 'john', 'doe']))
```

```python
"abc123".islower()   # islower() checks if ALL the letters in the string are lowercase, digits like 1,2,3 are just ignored/skipped during this check
# since "abc" are all lowercase letters and there are no uppercase letters anywhere, this comes back True
```

```python
"abc123".isupper()   # isupper() checks if ALL the letters in the string are uppercase, digits are ignored again
# since "abc" is lowercase, not a single uppercase letter exists here, so this comes back False
```

```python
"abc123".isalpha()   # isalpha() checks if EVERY character is a letter, no digits, no spaces, no symbols allowed at all
# since this string has the digits 1,2,3 mixed in with the letters, this comes back False
```

```python
"abc123".isalnum()   # isalnum() checks if EVERY character is either a letter OR a number (alnum = alpha + numeric), no spaces or symbols allowed
# "abc" are letters and "123" are digits, nothing else is in there, so this comes back True
```

```python
"      ".isspace()   # isspace() checks if the string is made up of ONLY whitespace characters (spaces, tabs, newlines) and has at least one character
# this string is nothing but spaces, so this comes back True
```

```python
"123".isdigit()   # isdigit() checks if every character in the string is a digit (0-9)
# all three characters here are digits, so this comes back True
```

```python
"½".isnumeric()   # isnumeric() is a bit broader than isdigit(), it also says True for things like fractions, superscripts, and number characters from other scripts
# "½" is a number-like character (a fraction), so isdigit() would actually say False on this, but isnumeric() says True
```

# More String Methods
the methods above were the common ones, here are the rest of the built-in string methods that we hadn't touched yet, just so this notebook covers all of them

### changing the case of a string
`upper()`, `lower()`, `capitalize()`, `title()`, `swapcase()`, `casefold()`

```python
sample = "Hello World FROM python"

print(sample.upper())       # upper() turns every single letter into UPPERCASE, no matter how it was typed originally
print(sample.lower())       # lower() turns every single letter into lowercase, the exact opposite of upper()
print(sample.capitalize())  # capitalize() makes ONLY the very first letter of the whole string uppercase, and forces every other letter to lowercase
print(sample.title())       # title() makes the first letter of EVERY word uppercase (like a book title), and lowercases the rest of each word
print(sample.swapcase())    # swapcase() flips the case of every letter, whatever was upper becomes lower and whatever was lower becomes upper
```

```python
# casefold() looks a lot like lower(), and for normal english text they behave the same
# the difference only shows up with some special characters from other languages, casefold() is more aggressive and is meant specifically for comparing text while ignoring case
germanWord = "straße"
print(germanWord.lower())     # lower() leaves the special "ß" character as it is
print(germanWord.casefold())  # casefold() converts "ß" into "ss" so that it can be safely compared against "strasse"
```

### finding and counting things inside a string
`find()`, `rfind()`, `count()`, `startswith()`, `endswith()`

```python
# find() searches for a piece of text and gives back the position (index) where it FIRST shows up, counting from 0
print(txt.find("name"))

# find() is a lot like index() from earlier, but it behaves nicer when the text is missing
print(txt.find("xyz"))     # since "xyz" does not exist anywhere in txt, find() quietly gives back -1 instead of crashing
# print(txt.index("xyz"))  # doing the same search with index() instead would actually throw an error, that's the main difference between the two
```

```python
# rfind() is the same as find(), but it searches starting from the END of the string, so it gives you the position of the LAST match instead of the first
print(txt.rfind("o"))
```

```python
# count() tells you how many times a piece of text shows up inside the string
print(txt.count("o"))   # counting how many times the letter "o" shows up in the whole sentence
print(txt.count("hello world my name is john doe"))
```

```python
# startswith() and endswith() just ask a simple yes/no question and give back True or False
print(txt.startswith("hello"))  # does the string begin with "hello"? yes, so True
print(txt.startswith("world"))  # does it begin with "world"? no, so False
print(txt.endswith("doe"))      # does the string end with "doe"? yes, so True
print(txt.endswith("john"))     # does it end with "john"? no, because "doe" comes after it, so False
```

### lining text up with padding
`ljust()`, `rjust()`
(we already saw `center()` and `zfill()` earlier, these two work the same way but push the text to one side instead of the middle)

```python
word = "cat"

print(word.ljust(10, "*"))   # ljust() keeps the text on the LEFT and fills up the remaining space on the right with the fill character we gave it
print(word.rjust(10, "*"))   # rjust() keeps the text on the RIGHT and fills up the remaining space on the left instead
# if you don't pass a fill character, both of these just use plain spaces to fill the gap
```

### checking what kind of characters are inside a string
we already covered `islower()`, `isupper()`, `isalpha()`, `isalnum()`, `isspace()`, `isdigit()` and `isnumeric()` above, here are the remaining ones

```python
print("hello123".isascii())   # isascii() checks if every character belongs to the basic ascii set (plain english letters, digits, common symbols)
print("héllo".isascii())      # the accented "é" is not a plain ascii character, so this comes back False
```

```python
print("123".isdecimal())   # isdecimal() checks if every character is a plain base-10 digit (0-9), this is the strictest of the three "number checking" methods
print("½".isdecimal())     # unlike isnumeric(), isdecimal() does NOT count fraction characters like "½" as valid, so this is False
# quick way to remember the difference: isdecimal() is strictest, isdigit() is a bit wider, isnumeric() is the widest of the three
```

```python
print("john_doe".isidentifier())  # isidentifier() checks if the string COULD legally be used as a variable name in python
print("john doe".isidentifier())  # this has a space in it, and variable names are not allowed to have spaces, so this comes back False
print("2ndPlace".isidentifier())  # variable names also can't start with a number, so this is False too
```

```python
print("Hello World".istitle())   # istitle() checks if the string is ALREADY in "title case", meaning every word starts with a capital letter and the rest of that word is lowercase
print("Hello world".istitle())   # here only the first word is capitalized properly, the second word "world" is not, so this comes back False
```

```python
print("hello world".isprintable())    # isprintable() checks if every character in the string can actually be displayed/printed on screen
print("hello\tworld".isprintable())   # \t is a tab character, and tabs count as "non-printable" for this check, so this comes back False
```

### finding and replacing pieces of text
`replace()`, `rpartition()`

```python
# replace() swaps out every occurrence of one piece of text with another piece of text
print(txt.replace("john doe", "jane doe"))
print(txt.replace("o", "0"))   # replacing every single "o" in the sentence with the digit "0"
```

```python
# rpartition() works exactly like partition() from earlier, but it looks for the LAST occurrence of the separator instead of the first one
print(txt.partition("o"))    # splits around the FIRST "o" it finds
print(txt.rpartition("o"))   # splits around the LAST "o" it finds
```

### a few extra ones
`encode()`, `expandtabs()`, `maketrans()` + `translate()`

```python
# computers don't actually store letters, they store numbers, encode() converts our readable text into raw bytes using a chosen encoding scheme (utf-8 is the most common one)
encoded = "hello".encode("utf-8")
print(encoded)
print(type(encoded))   # notice the type is "bytes", not "str" anymore, this is a completely different data type meant for storage/sending over a network
```

```python
# expandtabs() replaces every tab character (\t) in the string with a fixed number of spaces, useful for making tabbed text line up consistently everywhere
tabbedText = "name\tage\tcity"
print(tabbedText)                     # this is how it looks with the raw tab characters
print(tabbedText.expandtabs(10))      # this replaces each tab with enough spaces to reach the next multiple of 10 columns
```

```python
# maketrans() and translate() are a team, they work together to swap multiple characters in one go
# maketrans() builds a "translation map" (behind the scenes it's just a dictionary of character-to-character swaps)
swapMap = str.maketrans("aeiou", "12345")   # every "a" becomes "1", every "e" becomes "2", and so on

# translate() then actually applies that map to a string, swapping every matching character all at once
print("hello world".translate(swapMap))
```
