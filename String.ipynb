# Str (string) 
sequence of character(`Alphabets, Numbers, Special symbols or Whitespace`) enclosed in single `' '` or double `" "` quotes

```python
# 345 is in quotes " ". So Python sees it as text (a string), not a number.
x = "345"

# type() of "123" shows str (because of the quotes)
print(type("123"))

# type() of x shows str too
print(type(x))
```

#### `len()` function is used to get the number of characters in a string

```python
# len() counts the characters. Spaces count too.
# "john das" has 8 characters (including the space).
print(len(myName))
```

##### Note : Value collected using the input function is of type `Str` by default, but can be converted into supported data type

using function like str, float, int , list, tuple, set, ... etc

```python
# input() always gives text (a string), even if the user types a number.
# int() changes that text into a real whole number.
n = int(input("Enter a number : "))

# It shows the value and its type. The type is int now.
print(n, type(n))

# n is a number now, so n + n is real addition (23 + 23 = 46).
# Without int(), it would join two texts together.
print(n + n)
```

```python
# STRING METHODS
# Methods are ready-made tools for strings.
# Write them like this: txt.something()
```

```python
# One sentence. We use it in the examples below.
txt = "hello world my name is john doe"
```

```python
print(txt.index("o"))   # index() gives the position of the FIRST "o" (counting from 0)
print(txt.rindex("o"))  # rindex() gives the position of the LAST "o"
```

```python
# strip() removes spaces at the start and the end. Spaces in the middle stay.
# It does NOT change the old string. It gives a new one. (Strings can not be changed.)
"       rohan sharma     ".strip()  # remove leading and trailing
```

```python
# lstrip() removes spaces only on the LEFT (start).
"       rohan sharma     ".lstrip()  # remove leading whitespace
```

```python
# rstrip() removes spaces only on the RIGHT (end).
"       rohan sharma     ".rstrip()  # remove trailing whitespace
```

```python
# txt is the same as before. No method changed it.
# Every string method gives a NEW string. The old one stays.
txt
```

```python
# split() cuts the sentence into a list of words.
# By default it cuts at spaces (also tab and new line).
# splits the string into a list (on whitespace by default)
print(txt.split())
```

```python
# You can choose which letter to cut at.
# splits the string into a list (on whitespace by default)
print(txt.split("a"))

# maxsplit limits how many cuts.
# maxsplit=1 means cut only ONE time, at the first "m".
print(txt.split("m" , maxsplit=1))
```

```python
# format() puts values into the {} spaces, in order.
print("hello world my name is {} and i am an {}".format("john" , "alien"))
```

```python
# format_map() is like format(), but we give a dictionary.
# Each {key} gets the value of that key.
info = {"name": "john", "creature": "alien"}
print("hello world my name is {name} and i am an {creature}".format_map(info))
```

```python
# splits the string into a list (on whitespace by default)
print(txt.split("m" , maxsplit=1))   # split() counts from the LEFT
print(txt.rsplit("m" , maxsplit=1))  # rsplit() counts from the RIGHT. It cuts the LAST "m".
```

```python
# removeprefix() removes text only if it is at the START.
# Otherwise nothing changes.
print(txt.removeprefix("hello"))  # "hello" is at the start, so it is removed
print(txt.removeprefix("john"))   # "john" is not at the start, so nothing is removed

# removesuffix() is the same, but for the END.
print(txt.removesuffix("hello"))  # "hello" is not at the end, so nothing changes
print(txt.removesuffix("doe"))    # "doe" is at the end, so it is removed
```

```python
print(txt.split("my"))       # split() cuts at EVERY "my" and removes "my"
print(txt.partition("my"))   # partition() cuts at the FIRST "my" only.
# It gives a tuple of 3: (before, "my", after)

# difference between split and partition
# split gives many pieces and removes the separator.
# partition always gives 3 pieces and keeps the separator in the middle.
```

```python
print(len(txt))              # the length of our sentence
print(txt.center(41))        # center() adds spaces on both sides to make the length 41
print(txt.center(41 , "-"))  # new length (more than the actual length) ,  fill char
# same, but it fills with "-" instead of spaces
```

```python
print(txt.zfill(41)) # new length (more than the actual length
# zfill() adds zeros "0" on the LEFT only.
# It is used for numbers in text, like "7" -> "007".
```

```python
# join() is the opposite of split(). It joins a list into one string.
# The string before .join() is the glue between the items.
print(("-").join(['hello', 'world', 'my', 'name', 'is', 'john', 'doe']))
```

```python
"abc123".islower()   # islower() checks: are ALL letters lowercase? Digits are ignored.
# "abc" is lowercase, so True
```

# More String Methods
The methods above are the common ones. Here are the rest.

### changing the case of a string
`upper()`, `lower()`, `capitalize()`, `title()`, `swapcase()`, `casefold()`

```python
sample = "Hello World FROM python"

print(sample.upper())       # upper() makes ALL letters UPPERCASE
print(sample.lower())       # lower() makes ALL letters lowercase
print(sample.capitalize())  # capitalize() makes ONLY the first letter big.
# All other letters become small.
print(sample.title())       # title() makes the first letter of EVERY word big.
# (like a book title)
print(sample.swapcase())    # swapcase() flips each letter.
# Big becomes small. Small becomes big.
```

```python
# casefold() is like lower(). For English they are the same.
# The difference shows in some special letters from other languages.
# casefold() is stronger. It is used to compare text without case.
germanWord = "straße"
print(germanWord.lower())     # lower() keeps "ß"
print(germanWord.casefold())  # casefold() changes "ß" to "ss"
```

### finding and counting things inside a string
`find()`, `rfind()`, `count()`, `startswith()`, `endswith()`

```python
# find() gives the position where the text FIRST shows up (from 0).
print(txt.find("name"))

# find() is like index(), but safer when the text is missing.
print(txt.find("xyz"))     # "xyz" is not there, so find() gives -1 (no error)
# print(txt.index("xyz"))  # index() would give an error.
# That is the main difference.
```

```python
# rfind() is like find(), but it starts from the END.
# It gives the position of the LAST match.
print(txt.rfind("o"))
```

```python
# count() tells how many times a text is in the string.
print(txt.count("o"))   # how many "o" in the sentence
print(txt.count("hello world my name is john doe"))
```

```python
# startswith() and endswith() answer yes or no (True or False).
print(txt.startswith("hello"))  # starts with "hello"? Yes, True
print(txt.startswith("world"))  # starts with "world"? No, False
print(txt.endswith("doe"))      # ends with "doe"? Yes, True
print(txt.endswith("john"))     # ends with "john"? No, False
```

### lining text up with padding
`ljust()`, `rjust()`
(We saw `center()` and `zfill()` before. These two push the text to one side.)

```python
word = "cat"

print(word.ljust(10, "*"))   # ljust() keeps the text on the LEFT.
# It fills the right side with the fill character.
print(word.rjust(10, "*"))   # rjust() keeps the text on the RIGHT.
# It fills the left side.
# Without a fill character, both use spaces.
```

### checking what kind of characters are inside a string
We saw `islower()`, `isupper()`, `isalpha()`, `isalnum()`, `isspace()`, `isdigit()` and `isnumeric()`. Here are the rest.

```python
print("hello123".isascii())   # isascii() checks: are all characters basic English letters, digits and symbols?
print("héllo".isascii())      # "é" is not basic, so False
```

```python
print("123".isdecimal())   # isdecimal() checks: is every character a normal digit (0-9)?
# It is the strictest of the three number checks.
print("½".isdecimal())     # "½" is not a normal digit, so False
# Easy way to remember: isdecimal() is the strictest. isdigit() is wider. isnumeric() is the widest.
```

```python
print("john_doe".isidentifier())  # isidentifier() checks: can this be a variable name in Python? True
print("john doe".isidentifier())  # it has a space, so False
print("2ndPlace".isidentifier())  # it starts with a number, so False
```

```python
print("Hello World".istitle())   # istitle() checks: does every word start big and the rest small?
print("Hello world".istitle())   # "world" starts small, so False
```

```python
print("hello world".isprintable())    # isprintable() checks: can every character show on the screen?
print("hello\tworld".isprintable())   # \t is a tab. A tab is not printable, so False
```

### finding and replacing pieces of text
`replace()`, `rpartition()`

```python
# replace() changes every match to new text.
print(txt.replace("john doe", "jane doe"))
print(txt.replace("o", "0"))   # every "o" becomes the digit "0"
```

```python
# rpartition() is like partition(), but it uses the LAST match.
print(txt.partition("o"))    # cuts at the FIRST "o"
print(txt.rpartition("o"))   # cuts at the LAST "o"
```

### a few extra ones
`encode()`, `expandtabs()`, `maketrans()` + `translate()`

```python
# Computers store numbers, not letters.
# encode() changes text into bytes. utf-8 is the most common way.
encoded = "hello".encode("utf-8")
print(encoded)
print(type(encoded))   # the type is "bytes", not "str".
# Bytes are used to save or send data.
```

```python
# expandtabs() changes each tab (\t) into spaces.
# It helps text line up.
tabbedText = "name\tage\tcity"
print(tabbedText)                     # with real tabs
print(tabbedText.expandtabs(10))      # each tab becomes spaces up to the next multiple of 10
```

```python
# maketrans() and translate() work together. They swap many letters at once.
# maketrans() makes a swap map (like a small dictionary).
swapMap = str.maketrans("aeiou", "12345")   # "a" becomes "1", "e" becomes "2", and so on

# translate() uses the map on a string. All matching letters change at once.
print("hello world".translate(swapMap))
```
