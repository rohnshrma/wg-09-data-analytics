# Dictionary
`{ }` are used to create a dictionary where properties are stored in a `key: value` pair format

-  ordered, unindexed, changable (mutable) and no duplicates are allowed

**Simple idea:** think of a dictionary like your phone contacts. The NAME is the key and the PHONE NUMBER is the value. You search by name, not by position number.

```python
# We make a dictionary called "car". It is like a small information card about one car.
# Curly brackets { } tell Python "this is a dictionary".
# Each line is one pair written as  key : value  (left side = label, right side = information).
car = {
    "brand" : "hyundai",   # key is "brand", value is "hyundai"
    "model" : "verna",     # key is "model", value is "verna"
    "year" : 2020,         # key is "year", value is the number 2020 (numbers do not need quotes)
    "availableColors" : ["red", "grey"],  # a value can also be a list of many things
    "year" : 2022,         # "year" is written again! A key can not repeat, so Python keeps the LAST one (2022) and forgets 2020
}
```

```python
# type() tells us what kind of thing this is. Here it will show <class 'dict'>, dict means dictionary.
print(type(car))
```

```python
# len() counts how many key : value pairs are inside.
# We wrote "year" two times, but it is counted only once, so the answer is 4 (not 5).
print(len(car))
```

```python
# print() shows the whole dictionary on screen.
# You will see year is 2022 only. The old 2020 is gone because duplicates are not allowed.
print(car) # ordered and no duplicates are allowed
```

```python
# This line gives an ERROR on purpose. Do not worry, it is part of learning.
# A list uses position numbers (0, 1, 2), but a dictionary does not.
# There is no key called 0 in car, so Python says KeyError.
print(car[0]) # unindexed
```

```python
# To get a value, write the dictionary name and put the KEY inside square brackets [ ].
# values can be accessed using "keys"
print(car["model"])           # shows: verna
print(car["availableColors"]) # shows the full list: ['red', 'grey']
```

```python
# To change a value, use the same key and put a new value after the = sign.
# The old value 2022 is replaced by 2032. Nothing is printed here, it just changes silently.
car["year"] = 2032 # changable
```

```python
# Let us check that the change really happened. It will show 2032.
print(car["year"])
```

```python
# A key does not have to be a word. It can be a number too.
# Here the number is the key (like a roll number) and the name is the value.
marks = {
    0 : "ravi",
    1 : "savita",
    2 : "manju"
}
```

```python
# Print the full dictionary first.
print(marks)
# Now print one by one. Remember: 0, 1, 2 here are KEYS (labels we chose), not positions.
# It looks like a list, but it is not.
print(marks[0]) # ravi
print(marks[1]) # savita
print(marks[2]) # manju
```

```python
# If we use a key that is NOT in the dictionary, Python does not give an error here.
# It simply adds a brand new key : value pair at the end.
# changing value of key which doesn't exists will result in formation of new property
car["isAvailable"] = False # False (capital F, no quotes) means no
```

```python
# Print again to see the new pair "isAvailable" : False was added at the end.
print(car)
```

```python
# Dictionary Methods
# A method is a ready-made tool that Python gives us. We use it by writing  name.tool()
# Below we will try the most useful tools one by one.
```

```python
# keys() gives us only the labels (the left side) of the dictionary.
# returns list of keys
car.keys()
```

```python
# values() gives us only the information (the right side) of the dictionary.
# returns list of values
car.values()
```

```python
# items() gives us both together. Each pair comes as a small group like ("brand", "hyundai").
# That small group is called a tuple.
# returns a list containing each key : value pair as a tuple
car.items()
```

```python
# This gives an ERROR on purpose. "mileage" is not in car yet.
# Using square brackets [ ] with a missing key crashes the program with KeyError.
# raises an error if key not found
car["mileage"]
```

```python
# get() is the safe way to read a value. It never crashes.
# returns the value of the key specified, if not found returns None
print(car.get("brand"))   # key exists, so it shows: hyundai
print(car.get("mileage")) # key is missing, so it shows None (None means "nothing is here")
```

```python
# pop() deletes the key you name and gives you back its value.
# Here the pair "isAvailable" : False is removed from car.
# removes the value of the key specified
car.pop("isAvailable") 
```

```python
# popitem() deletes the LAST pair that was added, you do not need to say which one.
print(car)       # before deleting
car.popitem()    # the last pair is removed
print(car)       # after deleting, the last pair is gone
```

```python
# update() changes or adds many pairs in one go. Give it a dictionary inside ( ).
# "year" already exists, so its value changes to 2020.
# "mileage" does not exist, so it is added as a new pair.
# can set value of a existing key or add a new key:value pair
car.update({"year" : 2020 , "mileage" : "10 kmpl"})
```

```python
# Writing only the name in the last line of a cell shows its value, so we can check the result.
car
```

```python
# copy() makes a second, separate dictionary with the same data.
# Now "carrr" is a photocopy of car. Changing one will not change the other.
carrr = car.copy()
```

```python
# Check the copy. It has the same data as car.
carrr
```

```python
# clear() removes everything inside. The dictionary stays but becomes empty { }.
# Only the copy (carrr) is emptied, the original car is safe.
carrr.clear()
```

```python
# Check again. It shows empty curly brackets {}.
carrr
```

```python
# fromkeys() builds a NEW dictionary from a list of keys.
# Every key gets the same starting value, here 100 (like giving 100 marks to everybody at the start).
# takes a list of keys and a default value and returns a new dictionary
x = car.fromkeys(["prashant" , "sandhya", "ram"] , 100 )
```

```python
# Check the new dictionary x. It shows {'prashant': 100, 'sandhya': 100, 'ram': 100}
```

```python
# setdefault() looks for the key "vinay".
# If the key is missing, it adds "vinay" with the value 1000000 and shows that value.
# If the key was already there, it keeps the old value and changes nothing.
x.setdefault("vinay" , 1000000 )
```

```python
# Print x to see that "vinay" is now added.
print(x)
```
