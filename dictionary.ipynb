# Dictionary
`{ }` are used to create a dictionary where properties are stored in a `key: value` pair format

-  ordered, unindexed, changable (mutable) and no duplicates are allowed

**Easy idea:** a dictionary is like your phone contacts. The name is the key. The phone number is the value.

```python
# This is a dictionary named "car". { } make a dictionary.
# Each line is key : value.
car = {
    "brand" : "hyundai",   # key = "brand", value = "hyundai"
    "model" : "verna",     # key = "model", value = "verna"
    "year" : 2020,         # numbers do not need quotes
    "availableColors" : ["red", "grey"],  # a value can be a list
    "year" : 2022,         # same key again! Python keeps the last one (2022)
}
```

```python
# type() shows what this is. Answer: dict (dictionary).
print(type(car))
```

```python
# len() counts the pairs. Answer: 4 (the repeated "year" counts once).
print(len(car))
```

```python
# Show the whole dictionary. "year" is 2022 only.
print(car) # ordered and no duplicates are allowed
```

```python
# This gives an ERROR on purpose.
# A dictionary has no position numbers. There is no key 0.
print(car[0]) # unindexed
```

```python
# Use the key inside [ ] to get the value.
# values can be accessed using "keys"
print(car["model"])           # verna
print(car["availableColors"]) # ['red', 'grey']
```

```python
# Use the key to change the value. 2022 becomes 2032.
car["year"] = 2032 # changable
```

```python
# Check the new value. It shows 2032.
print(car["year"])
```

```python
# A key can be a number too.
marks = {
    0 : "ravi",
    1 : "savita",
    2 : "manju"
}
```

```python
# Show all of it.
print(marks)
# 0, 1, 2 are keys here, not positions.
print(marks[0]) # ravi
print(marks[1]) # savita
print(marks[2]) # manju
```

```python
# A new key is not an error. Python adds it.
# changing value of key which doesn't exists will result in formation of new property
car["isAvailable"] = False # False means no
```

```python
# Check. "isAvailable" is added at the end.
print(car)
```

```python
# Dictionary Methods
# A method is a ready tool. Write it like this: name.tool()
```

```python
# keys() gives all the keys (names).
# returns list of keys
car.keys()
```

```python
# values() gives all the values.
# returns list of values
car.values()
```

```python
# items() gives each key and value together as a pair.
# returns a list containing each key : value pair as a tuple
car.items()
```

```python
# ERROR on purpose. "mileage" is not in car.
# [ ] with a missing key gives KeyError.
# raises an error if key not found
car["mileage"]
```

```python
# get() is safe. It gives no error.
# returns the value of the key specified, if not found returns None
print(car.get("brand"))   # hyundai
print(car.get("mileage")) # None (means nothing)
```

```python
# pop() removes the key you say.
# removes the value of the key specified
car.pop("isAvailable") 
```

```python
# popitem() removes the last pair.
print(car)       # before
car.popitem()    # remove last pair
print(car)       # after
```

```python
# update() changes or adds many pairs at once.
# "year" changes to 2020. "mileage" is new, so it is added.
# can set value of a existing key or add a new key:value pair
car.update({"year" : 2020 , "mileage" : "10 kmpl"})
```

```python
# Check the result.
car
```

```python
# copy() makes a second dictionary with the same data.
carrr = car.copy()
```

```python
# Check the copy.
carrr
```

```python
# clear() removes everything. Only the copy is emptied.
carrr.clear()
```

```python
# It is empty now: {}
carrr
```

```python
# fromkeys() makes a new dictionary from a list of keys.
# Every key gets the same value, 100.
# takes a list of keys and a default value and returns a new dictionary
x = car.fromkeys(["prashant" , "sandhya", "ram"] , 100 )
```

```python
# Check x. Every name has 100.
```

```python
# setdefault() adds "vinay" only if it is not there.
# If it is already there, nothing changes.
x.setdefault("vinay" , 1000000 )
```

```python
# Check. "vinay" is added.
print(x)
```
