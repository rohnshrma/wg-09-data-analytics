### How the game works
- The board is a 3×3 grid with positions numbered **1 to 9**:
```
 1 | 2 | 3
---+---+---
 4 | 5 | 6
---+---+---
 7 | 8 | 9
```
- Player **X** goes first, then **O**, taking turns.
- On your turn, type a number 1-9 to place your mark in that box.
- The first player to get **3 in a row** (across, down or diagonally) wins.
- If all 9 boxes are full and nobody has won, it is a **draw**.
