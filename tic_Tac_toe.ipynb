- The board is a 3×3 grid with positions numbered **1 to 9**:

## Task 12 - Put it all together (the game loop)
Now combine Tasks 1-11 into **one cell** that plays a full game:

1. Set up the board and variables.
2. Use a **`while` loop** that keeps going while there is **no winner** and the game is **not a draw**.
3. Inside the loop, each turn:
   - Show the board
   - Ask the current player for a move
   - If the move is invalid → show a message and `continue` (same player tries again)
   - Place the mark and count the move
   - Check for a winner → if found, stop the loop
   - Check for a draw → if yes, stop the loop
   - Switch players
4. After the loop, show the final board and print the result:
   - `Player X wins! 🎉` / `Player O wins! 🎉`, or
   - `It's a draw! 🤝`

💡 *Think:* Should you switch players **before** or **after** checking for a winner? What happens to the winner message if you get the order wrong?

✅ *Test these games:*
- X wins across a row
- O wins down a column
- Someone wins on a diagonal
- A full-board draw
- Typing letters, `0`, `10`, and an already-taken box

````python
# ============================================================
# TIC TAC TOE - FULL GAME (Task 12)
# Think of this like a board game night: we set up the board,
# then keep taking turns until someone wins or the board is full.
# ============================================================


# ---------- STEP 1: SET UP THE GAME ----------

# The board is a list (like a row of 9 little boxes).
# [" "] * 9 means "repeat one blank space 9 times" = 9 empty boxes.
# Box numbers inside Python start from 0, so the boxes are 0,1,2,...,8.
board = [" "] * 9

# Whose turn is it? X always starts. (A variable is just a label on a box
# that remembers a value.)
current_player = "X"

# Who won? Empty text "" means "nobody has won yet".
winner = ""

# A counter for how many moves have been played. Starts at 0.
# A full board = 9 moves, so this helps us detect a draw.
moves = 0

# All 8 ways to win, written as "which 3 boxes make a line".
# This is a list of lists (a list that contains smaller lists).
# Remember the index map:   0 | 1 | 2
#                           3 | 4 | 5
#                           6 | 7 | 8
winning_lines = [
    [0, 1, 2],   # top row
    [3, 4, 5],   # middle row
    [6, 7, 8],   # bottom row
    [0, 3, 6],   # left column
    [1, 4, 7],   # middle column
    [2, 5, 8],   # right column
    [0, 4, 8],   # diagonal: top-left to bottom-right
    [2, 4, 6]    # diagonal: top-right to bottom-left
]


# ---------- STEP 2: THE GAME LOOP ----------

# A while loop = "keep repeating as long as the condition is True".
# Here: keep playing while NOBODY has won (winner == "")
#       AND the board isn't full yet (moves < 9).
# The moment someone wins or moves hits 9, the loop stops.
while winner == "" and moves < 9:

    # ----- 2a. Show the board to the players -----
    print()  # empty line, just to make the output look neat
    # f-string: the f before the quotes lets us drop variables into
    # the text using { }. So {board[0]} becomes whatever is in box 0.
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")   # the horizontal divider line
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()

    # ----- 2b. Ask the current player where they want to play -----
    # input() pauses the program and waits for the user to type.
    # IMPORTANT: whatever they type always comes back as TEXT (a string),
    # even if they type a number like 5.
    move = input(f"Player {current_player}, enter position (1-9):")

    # ----- 2c. Check #1: did they type a number at all? -----
    # .isdigit() is True only if the text is made of digits (like "7").
    # "abc" or "" or "-3" would be False.
    # "not" flips it: "if it is NOT a number...".
    if not move.isdigit():
        print("Please enter a number between 1 to 9.")
        # continue = "skip the rest of this round and jump back to the
        # top of the loop". The player doesn't lose their turn;
        # they just get asked again.
        continue

    # Now we know it's made of digits, so convert text -> real number.
    # "5" (text) becomes 5 (number), so we can compare and do maths.
    move = int(move)

    # ----- 2d. Check #2: is the number between 1 and 9? -----
    # "or" means: if EITHER side is True, the whole thing is True.
    # So: too small (like 0) OR too big (like 10) = invalid.
    if move < 1 or move > 9:
        print("Position must be between 1-9.")
        continue  # ask the same player again

    # Players think in 1-9, but Python lists count from 0.
    # So position 1 is box 0, position 5 is box 4, etc. Subtract 1.
    index = move - 1

    # ----- 2e. Check #3: is that box still empty? -----
    # != means "is not equal to". If the box isn't a blank space,
    # someone already played there.
    if board[index] != " ":
        print("That board is already taken.")
        continue  # ask the same player again

    # ----- 2f. The move is valid, so place the mark -----
    # Put "X" or "O" into the chosen box.
    board[index] = current_player
    # Add 1 to the move counter (same as moves = moves + 1).
    moves += 1

    # ----- 2g. Did this move win the game? -----
    # A for loop goes through the list one item at a time.
    # Each "line" here is one of the 8 winning combos, e.g. [0, 1, 2].
    for line in winning_lines:
        # Pull out the three box numbers of this line.
        a = line[0]
        b = line[1]
        c = line[2]

        # A line wins if:
        #   - box a is not empty (otherwise 3 blanks would count as a "win"!)
        #   - AND box a matches box b
        #   - AND box b matches box c
        # "and" means ALL parts must be True.
        if board[a] != " " and board[a] == board[b] and board[b] == board[c]:
            winner = board[a]   # remember who has 3 in a row ("X" or "O")
            break               # break = stop the for loop early; no need to check more lines

    # ----- 2h. Switch to the other player -----
    # We only swap if nobody has won yet. Once someone wins, the game is
    # over, so we keep current_player as it is (the winner) and don't pass the turn.
    if winner == "":
        if current_player == "X":
            current_player = "O"
        else:
            current_player = "X"
    # After this, Python goes back to the top of the while loop
    # and checks the condition again.


# ---------- STEP 3: THE GAME IS OVER - SHOW THE RESULT ----------

# We reach here only after the while loop ends
# (either someone won or the board filled up).
# Show the final board one last time.
print()
print(f" {board[0]} | {board[1]} | {board[2]} ")
print("---+---+---")
print(f" {board[3]} | {board[4]} | {board[5]} ")
print("---+---+---")
print(f" {board[6]} | {board[7]} | {board[8]} ")
print()

# If winner is still empty, the loop ended because the board filled up
# with no 3-in-a-row, so it's a draw.
if winner == "":
    print("It's a draw!")
else:
    # Otherwise announce whoever is stored in `winner`.
    print(f"Player {winner} wins!")
````

````python
board = [" "] * 9

current_player = "X"
winner = ""
moves = 0

winning_lines = [
    [0, 1, 2],
    [3, 4, 5],
    [6, 7, 8],
    [0, 3, 6],
    [1, 4, 7],
    [2, 5, 8],
    [0, 4, 8],
    [2, 4, 6]
]

while winner == "" and moves < 9:

    print()
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()

    move = input(f"Player {current_player}, enter position (1-9):")

    if not move.isdigit():
        print("Please enter a number between 1 to 9.")
        continue

    move = int(move)

    if move < 1 or move > 9:
        print("Position must be between 1-9.")
        continue

    index = move - 1

    if board[index] != " ":
        print("That board is already taken.")
        continue

    board[index] = current_player
    moves += 1

    for line in winning_lines:
        a = line[0]
        b = line[1]
        c = line[2]
        if board[a] != " " and board[a] == board[b] and board[b] == board[c]:
            winner = board[a]
            break

    if winner == "":
        if current_player == "X":
            current_player = "O"
        else:
            current_player = "X"

print()
print(f" {board[0]} | {board[1]} | {board[2]} ")
print("---+---+---")
print(f" {board[3]} | {board[4]} | {board[5]} ")
print("---+---+---")
print(f" {board[6]} | {board[7]} | {board[8]} ")
print()

if winner == "":
    print("It's a draw!")
else:
    print(f"Player {winner} wins!")
````
