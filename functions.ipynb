````python
# BMI calculator WITHOUT a function (runs once, top to bottom)

# input() asks the user to type something; it always gives back TEXT.
# int() turns that text into a whole number (e.g. "100" -> 100).
weight = int(input("Enter weight in kg's : "))
# float() turns text into a number with decimals (e.g. "1.90" -> 1.9).
height = float(input("Enter height in meter's : "))

# BMI formula: weight divided by (height x height).
# ** means "power of", so height ** 2 is height squared.
bmi = weight / height ** 2

# if / elif / else = a decision ladder. Python checks top to bottom
# and runs the FIRST block whose condition is True.
if bmi < 18.5:
    # f"...{bmi}..." (f-string) drops the value of bmi into the sentence.
    print(f"Underweight as your bmi is {bmi}")
elif bmi >= 18.5 and bmi < 25:      # "and" = both sides must be True
    print(f"Normal weight as your bmi is {bmi}")
# NOTE: this one is a plain "if", not "elif", so it starts a NEW ladder.
# That means for a normal BMI, the "else" below also runs and prints "Obese".
# Changing this "if" to "elif" fixes it.
if bmi >= 25 and bmi < 30:
    print(f"Overweight as your bmi is {bmi}")
else:
    print(f"Obese as your bmi is {bmi}")
````

# functions
reusable block of code , where the code inside of a function runs only when called by its name

---
`def` keyword is used to define a function followed by the function name


```python

# creating a function
def funcName( parameter(optional) ):
    # code to be executed

# calling a function
funcName(arguments)
```

````python
# "def" means "I am DEFINING (creating) a function".
# addition is the function's name; a and b are parameters:
# empty slots that will be filled when someone uses the function.
# Defining only saves the recipe - nothing runs yet.
def addition(a,b):
    # The indented line is the function's body (the work it does).
    # It adds the two slots and shows the answer on screen.
    print(a + b)
````

````python
# Calling the function = using the recipe.
# 12 and 12 are arguments: the real values that go into a and b.
addition(12,12)
````

````python
# a , b => parameters (variables without values)
# 12, 12 => arguments (values to parameters)
````

````python
# Same function, different values - that is the benefit of functions:
# write once, reuse as many times as you like.
addition(24,45)
````

````python
# The BMI code from above, packed inside a function with NO parameters.
# The empty brackets () mean it needs nothing passed in;
# it asks the user for the values itself using input().
def calcBmi():
    # Everything indented under def belongs to the function.
    weight = int(input("Enter weight in kg's : "))      # ask + convert text to whole number
    height = float(input("Enter height in meter's : ")) # ask + convert text to decimal number
    bmi = weight / height ** 2                           # BMI formula
    if bmi < 18.5:
        print(f"Underweight as your bmi is {bmi}")
    elif bmi >= 18.5 and bmi < 25:
        print(f"Normal weight as your bmi is {bmi}")
    if bmi >= 25 and bmi < 30:   # same plain "if" quirk as before (should be elif)
        print(f"Overweight as your bmi is {bmi}")
    else:
        print(f"Obese as your bmi is {bmi}")
````

````python
# Calling the function 3 times = 3 separate BMI checks.
# Each call asks for weight and height again.
calcBmi()
calcBmi()
calcBmi()
````

````python
# Better version: weight and height are PARAMETERS, so the caller
# hands the values in and the function doesn't need to ask with input().
def bmiCalculator(weight, height):
    bmi = weight / height ** 2      # formula, using the values handed in
    if bmi < 18.5:
        print(f"Underweight as your bmi is {bmi}")
    elif bmi >= 18.5 and bmi < 25:
        print(f"Normal weight as your bmi is {bmi}")
    if bmi >= 25 and bmi < 30:      # same plain "if" quirk (should be elif)
        print(f"Overweight as your bmi is {bmi}")
    else:
        print(f"Obese as your bmi is {bmi}")
````

````python
# 100 goes into weight, 1.80 goes into height (matched by order).
bmiCalculator(100, 1.80)
````

## Solved Example 1 - Even or odd counter
Function `countEvenOdd(n)` takes a number `n`, goes through every number from 1 to `n`, and prints how many are even and how many are odd.

````python
# n is the parameter: the last number we count up to.
def countEvenOdd(n):
    # Two counters (like tally marks), both start at 0.
    even = 0
    odd = 0

    # Loop visits 1, 2, 3 ... up to n. (range stops BEFORE the end, so n + 1.)
    for num in range(1, n + 1):
        # % gives the remainder after dividing. Even numbers leave 0 when divided by 2.
        if num % 2 == 0:
            even += 1      # even = even + 1
        else:
            odd += 1       # otherwise it must be odd

    # After the loop finishes, show the totals.
    print(f"Between 1 and {n}: {even} even, {odd} odd")

countEvenOdd(10)
countEvenOdd(7)
````

## Solved Example 2 - Grade report for a list of marks
Function `gradeReport(marks)` takes a list of marks and prints the grade for each (A: 80+, B: 60-79, C: 40-59, otherwise Fail), then prints how many students passed.

````python
# marks is the parameter: a list of numbers like [85, 52, 37].
def gradeReport(marks):
    passed = 0    # counter for students who passed

    # Go through the list one mark at a time.
    for m in marks:
        # Ladder: Python picks the first condition that is True.
        if m >= 80:
            grade = "A"
        elif m >= 60:
            grade = "B"
        elif m >= 40:
            grade = "C"
        else:
            grade = "Fail"

        print(f"Marks {m} -> {grade}")

        # Anything except "Fail" counts as a pass.
        if grade != "Fail":
            passed += 1

    print(f"{passed} out of {len(marks)} students passed")

gradeReport([85, 52, 37, 66, 91])
gradeReport([30, 45])
````
