# Number Guessing Game 🔢

**How the game works (in plain words):**
1. The computer secretly picks a number.
2. You try to guess it - you get 10 tries.
3. After every guess the computer gives a hint: *"Too high"* or *"Too low"*.
4. Guess it within 10 tries to win.

**Concepts used:** `import` & `random.randint`, variables, `while` loops, `input()` and `int()` (type conversion), comparison operators (`>`, `<`, `==`), `if / elif / else`, `break`, `+=`, f-strings and `print()`.

```python
# "import" = borrowing a ready-made tool from Python's toolbox.
# "random" is a toolbox (module) about chance/luck.
# "randint" is one tool inside it: it gives you a random WHOLE number
# between two numbers - like rolling a very big dice.
from random import randint
```

```python
# The computer secretly picks a random whole number and stores it in "secret".
# randint(1, 101) includes BOTH ends, so the number can be anything from 1 to 101.
# (Small catch: the game says 1-100, so randint(1, 100) would match exactly.)
secret = randint(1,101)
```

```python
# The player gets at most 10 tries.
max_attempts = 10

# A counter for how many guesses have been made so far. Starts at 0.
attempts = 0

# WHILE LOOP = "keep repeating this block as long as the condition is True".
# Here: keep asking for guesses while the player has used fewer than 10 tries.
while attempts < max_attempts:

    # Count this try. "attempts += 1" is short for "attempts = attempts + 1".
    attempts += 1

    # input() waits for the player to type something and press Enter.
    # Whatever is typed always comes in as TEXT (e.g. "42" not 42),
    # so int(...) converts that text into a real number we can compare.
    # (If the player types letters, int() will crash with a ValueError.)
    guess = int(input("Enter your guess (1-100) : "))

    # Compare the guess with the secret number:
    # ">" means "greater than" -> the guess is bigger than the secret.
    if guess  > secret:
        print("Too high! Try Low")
    # "elif" = "else if" -> checked only when the "if" above was False.
    # "<" means "less than" -> the guess is smaller than the secret.
    elif guess < secret:
        print("Too Low! Try High")
    # "else" = not bigger, not smaller... so it must be EQUAL. The player won!
    else:
        # f"..." is an f-string: values inside { } get filled in automatically.
        print(f"Congratulations! You've guessed the correct number {secret} in {attempts} attempts.")
        # "break" = stop the loop immediately, even if tries are left.
        break


# After the loop ends: if all 10 tries were used, the player (probably) lost,
# so we reveal the secret number.
# Small catch: if the player wins on EXACTLY the 10th try, attempts is also 10,
# so this message prints even after winning. Checking "guess != secret"
# as well would fix that.
if attempts == max_attempts:
    print(f"Failed to Guess the Correct Number : {secret}")
```
