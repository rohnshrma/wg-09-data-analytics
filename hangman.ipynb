# Hangman 🎯

**How the game works (in plain words):**
1. The computer secretly picks a word from a list.
2. You see blanks `_ _ _ _` - one blank per letter.
3. You guess one letter at a time.
   - Right letter → it appears in its blank(s).
   - Wrong letter → you lose one of your 10 lives.
4. Fill every blank before your lives run out to win.

**Concepts used:** `import` & `random.choice`, lists, variables, `len()`, list repetition (`["_"] * n`), `while` loops, `if / else`, `and` / `or` / `not`, `in` (membership), `continue`, `.append()`, `enumerate()`, string `.join()`, f-strings, `input()` and `print()`.

```python
# ============================================================
# HANGMAN GAME
# Idea: The computer secretly picks a word. You guess it one
# letter at a time. Every wrong letter costs you one "life".
# Guess all the letters before your lives run out and you win!
# ============================================================

# "import" = borrowing a ready-made tool from Python's toolbox.
# "random" is a toolbox (module) full of tools about chance/luck.
# "choice" is one tool inside it: give it a list, it picks ONE item at random
# (like pulling a chit out of a bowl with your eyes closed).
from random import choice

# A LIST of possible secret words. A list = a row of boxes holding values,
# written inside square brackets [ ] and separated by commas.
words = ["cat", "bat", "spiderman", "lanterns" ,"hulk" ]

# The computer picks one word at random from the list and stores it
# in a variable called "word". A variable is just a labelled box that holds a value.
word = choice(words)

# An EMPTY list. We will drop every letter the player tries into it,
# so we can remember what has already been guessed.
guessed_letters = []

# The maximum number of WRONG guesses allowed (the player's "lives").
max_attempts = 10

# A counter for how many WRONG guesses have been made so far. Starts at 0.
attempts = 0

# Build the blank board the player sees, e.g. for "hulk" -> ['_', '_', '_', '_']
# len(word)  -> counts how many letters are in the word (hulk -> 4)
# ["_"] * 4  -> repeats the list 4 times -> ['_', '_', '_', '_']
# Each "_" is a hidden letter that will be revealed when guessed correctly.
display = ["_"] * len(word)

# WHILE LOOP = "keep repeating this block as long as the condition is True".
# We keep playing as long as BOTH things are true ("and" means both must be True):
#   1) attempts < max_attempts  -> the player still has lives left
#   2) "_" in display          -> there is still at least one hidden letter
# The moment either becomes False (out of lives OR word complete), the loop stops.
while attempts < max_attempts and "_" in display:

    # print() shows text on the screen.
    # f"..." is an f-string: anything inside { } gets replaced by its value.
    # "\n" means "new line" - it just adds an empty line for neatness.
    print(f"\nWord : {display}")

    # ", ".join(list) glues all items of a list into ONE string,
    # putting ", " between them. ['h', 'u'] -> "h, u"
    # (Using " quotes inside a " f-string only works in Python 3.12+.)
    print(f"Guessed Letters : {", ".join(guessed_letters)}")

    # Lives left = total lives minus wrong guesses used so far.
    print(f"Attempts Left : {max_attempts - (attempts) }")

    # input() pauses the program and waits for the player to type something
    # and press Enter. Whatever they type is saved (as text) in "guess".
    guess = input("\nEnter guess : ")

    # VALIDATION - checking the player typed something sensible.
    # len(guess) != 1      -> they typed more (or less) than one character
    # not guess.isalpha()  -> what they typed is NOT a letter (e.g. "5" or "@")
    # "or" means: if EITHER problem is true, the guess is invalid.
    if len(guess) != 1 or not guess.isalpha():
        print("\nGuess must be a single alphabet!")
        # "continue" = skip everything below and jump straight back to the
        # top of the loop for a fresh turn. (No life is lost for a typo.)
        continue


    # "in" checks membership: is this letter already inside our list?
    # If yes, tell the player and give them another go without penalty.
    if guess in guessed_letters:
        print("\nAlphabet already guessed! Try different letter.")
        continue

    # .append() adds a new item to the END of a list.
    # We remember this letter so it can't be guessed twice.
    guessed_letters.append(guess)

    # Is the guessed letter somewhere inside the secret word?
    if guess in word:
        # enumerate(word) walks through the word letter by letter and gives us
        # TWO things each time: the position number (i) and the letter (char).
        # For "hulk": (0,'h'), (1,'u'), (2,'l'), (3,'k')
        # Positions in Python start counting from 0, not 1.
        for i ,char in enumerate(word):
            # "==" asks "are these two equal?" (a single "=" means "store").
            # If the letter at this position matches the guess...
            if guess == char:
                # ...reveal it on the board by replacing "_" at the same position.
                # Using a loop means repeated letters (like both 'a's) all get revealed.
                display[i] = guess
        print("\nCorrect Guess")
    # "else" runs when the "if" above was False -> the letter is NOT in the word.
    else:
        print("\nWrong guess!")    
        # Wrong guess costs a life. "attempts += 1" is short for
        # "attempts = attempts + 1" (add one to the counter).
        attempts+=1



# ---- The loop has ended. Now we find out WHY it ended. ----
# If there are still "_" on the board, the player ran out of lives -> they lost.
if "_" in display:
    print(f"\nGame Over! Failed to Guess the word\nThe Word was {word}")
# Otherwise every letter was revealed -> they won!
else:
    print(f"Congratulations! you've guessed the correct word {word}")
```

```python
# DEMO of enumerate():
# enumerate() goes through something one item at a time and ALSO tells you
# the position (index) of each item, like a roll call: "0 - c, 1 - a, 2 - t".
# "i" gets the position number, "c" gets the letter.
for i , c in enumerate(word):
    # Print the letter first, then its position.
    print(c , i)
```
