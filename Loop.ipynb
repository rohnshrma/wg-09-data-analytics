# Loops:
A loop means "do the same work again and again" so you do not have to write the same line many times.
- while : fixed and non fixed iteration (it keeps going as long as a condition is True)
- for : fixed iteration (it goes through a known set of things, one by one)

#### iteration
One round of the loop. If a loop runs 10 times, it has 10 iterations.
#### iterable
Anything you can go through one item at a time, like a string ("hello"), a list ([1, 2, 3]) or range(5).

# While Loop fixed iteration
We know how many times it should run. It needs 3 things:
- initialization : the starting value (where to begin)
- condition : the question checked before every round. If True, the loop runs. If False, it stops
- update (increment / decrement) : change the value each round, or the loop never ends (infinite loop!)

```python
while condition:
    # code to be executed again and again while the condition is True
```

````python
i = 1 # initialization: start counting from 1. "i" is just a short name people use for a counter
while  i <= 10: # condition: keep going while i is 10 or less
    print(i) # show the current value of i on the screen
    i+=1 # update (increment): same as i = i + 1, so i grows by 1 every round. When i becomes 11, the condition is False and the loop stops
````

### Counting backwards (decrement)
Decrement means making the number smaller each round.

````python
i = 100 # start from 100 this time
while i >= 1: # keep going while i is 1 or more
    print(i) # show the current number
    i-=1 # decrement: same as i = i - 1, so the number goes down by 1 every round
````

### Order of the lines matters
In the next cell the update line is written BEFORE the print line. Run it and see how the output is different from the first example.

````python
i = 1 # start from 1
while i <= 10: # keep going while i is 10 or less
    i+=1 # first we add 1, so i becomes 2 before anything is shown
    print(i) # then we print. So the output starts at 2 and ends at 11 (not 1 to 10)
````

# While Loop with a string
A string is text. Every letter has a position number called an index. Counting starts from 0, not 1.

| letter | h | e | l | l | o |
|---|---|---|---|---|---|
| index | 0 | 1 | 2 | 3 | 4 |

````python
x = "hello" # a string (text) saved in the variable x
````

````python
# x[0] means "give me the letter at position 0". Here we do it by hand, one line for each letter
print(x[0]) # h
print(x[1]) # e
print(x[2]) # l
print(x[3]) # l
print(x[4]) # o
# This works, but it is boring to write. Imagine a word with 1000 letters! A loop does this job for us (next cell)
````

````python
i = 0 # index starts at 0 because the first letter is at position 0
while i < len(x): # len(x) counts the letters in x (5). We use "<" not "<=" because the last index is 4, not 5
    print(x[i]) # show the letter at the current position
    i+=1 # move to the next position
````

# While Loop with a list
A list holds many values in one variable. Just like a string, every value has an index starting from 0.

````python
marks = [12,23,34,34,55,4523,3423] # a list of numbers, written inside square brackets and separated by commas
````

````python
i = 0 # start from the first value (index 0)
while i < len(marks): # len(marks) counts the values in the list (7), so we stop before index 7
    print(marks[i]) # show the value at the current position
    i+=1 # go to the next value
````

# Sum of values (using a loop to calculate)
A loop can also collect things while it runs. Here we keep a running total.

````python
total = 0 # a box to keep the running total. It starts from 0 because we have not added anything yet
i = 0 # index starts at 0
while i < len(marks): # go through every value in the list
    total = total + marks[i] # take the old total, add the current value, and save it back in total
    i+=1 # move to the next value
print(total) # print only ONCE, after the loop is finished, to show the final answer
````

# For Loop fixed iteration
A for loop is the shorter and easier way when you already know what to go through. Python takes care of the counter for you, so no initialization, no condition and no update to write.

```python
for item in iterable:
    # code to be executed once for every item
```

````python
# range(1, 11) gives numbers from 1 up to 10. The last number (11) is NOT included
for i in range(1, 11): # i takes one number in each round: 1, then 2, then 3 ... until 10
    print(i) # show the current number
````

````python
# A for loop can go through a string directly. No index and no len() needed
for letter in x: # in each round, "letter" is the next character of "hello"
    print(letter) # show that character
````

````python
# A for loop can go through a list directly too
for mark in marks: # in each round, "mark" is the next value of the list
    print(mark) # show that value
````

## Completed examples (Intermediate)

### Example 1: Total bill (Intermediate task 1)

````python
# prices of the things a customer bought, in rupees
prices = [250, 120, 80, 450, 99]

total = 0 # bill starts at 0 before any item is added
i = 0 # start from the first item
while i < len(prices): # go through every price in the list
    print("Item", i + 1, "costs", prices[i]) # i + 1 because humans count items from 1, but the index starts at 0
    total = total + prices[i] # add this item price to the bill
    i+=1 # move to the next item

print("Total bill : ₹", total) # after the loop ends, show the final bill
````

### Example 2: Highest score without max() (Intermediate task 3)

````python
scores = [45, 78, 12, 91, 66, 34] # runs scored by a batsman in six matches

highest = scores[0] # assume the first score is the highest for now
for score in scores: # look at every score one by one
    if score > highest: # is this score bigger than the best one we have seen so far?
        highest = score # yes, so this becomes the new highest
print("Highest score :", highest) # after checking all scores, this is the real highest (91)
````

### Example 3: Password tries (Intermediate task 8)

````python
secret = "python123" # the correct password saved by us
tries = 0 # how many times the user has tried, starts at 0
logged_in = False # we will change this to True only if the password is right

while tries < 3 and not logged_in: # keep asking until 3 tries are used or the user gets it right
    guess = input("Enter password : ") # ask the user to type the password
    if guess == secret: # == checks if both are exactly same
        logged_in = True # correct, so this will stop the loop
    else:
        tries+=1 # wrong, so count one more try used
        print("Wrong password, tries left :", 3 - tries) # tell how many chances remain

if logged_in: # the loop is over, now check why it stopped
    print("Welcome")
else:
    print("Account locked")
````
